```python
%%writefile cross_tabulation_measures.py
"""Measures of association for one table of counts, as Cross Tabulation computes them.

SciPy and statsmodels give the chi-square, Fisher's test, McNemar and kappa, but not
Goodman-Kruskal's lambda and tau, the uncertainty coefficient, the ordinal pair
measures, the Cochran-Armitage trend test or the cumulative chi-square, so those are
written out here. Every function takes the counts as a rows x columns numpy array.

A table taken from one stratum of a layer can have an empty row or column; where a
ratio would then be 0/0, these return NaN (and a chi-square term with no expected count
adds nothing), which is what the app prints.
"""

import numpy as np
from scipy import stats
from scipy.stats import contingency

NAN = float("nan")


def _ratio(numerator, denominator):
    return numerator / denominator if denominator > 0 else NAN


# ---- chi-square tests ---------------------------------------------------------------

def expected_counts(counts):
    """(row total x column total) / n for every cell."""
    n = counts.sum()
    return np.outer(counts.sum(axis=1), counts.sum(axis=0)) / n if n > 0 else 0 * counts


def pearson(counts):
    """Pearson's chi-square: (statistic, df, P)."""
    expected = expected_counts(counts)
    with np.errstate(divide="ignore", invalid="ignore"):
        terms = np.where(expected > 0, (counts - expected) ** 2 / expected, 0.0)
    chi2 = terms.sum()
    df = (counts.shape[0] - 1) * (counts.shape[1] - 1)
    return chi2, df, stats.chi2.sf(chi2, df)


def yates(counts):
    """Yates' continuity correction (2x2 only): |O - E| shrunk by 0.5 before squaring."""
    expected = expected_counts(counts)
    difference = np.maximum(np.abs(counts - expected) - 0.5, 0.0)
    chi2 = (difference**2 / expected).sum()
    return chi2, 1, stats.chi2.sf(chi2, 1)


def likelihood_ratio(counts):
    """G2 = 2 * sum O ln(O / E); an empty cell adds nothing."""
    expected = expected_counts(counts)
    with np.errstate(divide="ignore", invalid="ignore"):
        terms = np.where(counts > 0, counts * np.log(counts / expected), 0.0)
    g2 = 2 * terms.sum()
    df = (counts.shape[0] - 1) * (counts.shape[1] - 1)
    return g2, df, stats.chi2.sf(g2, df)


def mantel_haenszel(tables):
    """The (Cochran-)Mantel-Haenszel correlation chi-square, 1 df: (statistic, P).

    Rows and columns are scored 1, 2, 3, ... in table order (SAS PROC FREQ's default, Mantel's equally spaced scores). With one table it is the Mantel-Haenszel
    chi-square of the chi-square tests, (n - 1) times the squared correlation of the scores;
    with one table per stratum it is the layer-adjusted CMH test.
    """
    numerator = variance = 0.0
    for counts in tables:
        n = counts.sum()
        if n < 2:
            continue
        rows, cols = counts.sum(axis=1), counts.sum(axis=0)
        x, y = np.arange(1.0, len(rows) + 1), np.arange(1.0, len(cols) + 1)
        observed = (x[:, None] * y[None, :] * counts).sum()
        expected = (x * rows).sum() * (y * cols).sum() / n
        var_x = (x**2 * rows).sum() - (x * rows).sum() ** 2 / n
        var_y = (y**2 * cols).sum() - (y * cols).sum() ** 2 / n
        numerator += observed - expected
        variance += var_x * var_y / (n - 1)
    if variance <= 0:
        return NAN, NAN
    statistic = numerator**2 / variance
    return statistic, stats.chi2.sf(statistic, 1)


# ---- nominal ------------------------------------------------------------------------

def phi(counts):
    """Phi: signed (ad - bc) / sqrt(margins) on a 2x2 table, sqrt(chi2 / n) otherwise."""
    if counts.shape == (2, 2):
        (a, b), (c, d) = counts
        margins = np.prod(counts.sum(axis=1)) * np.prod(counts.sum(axis=0))
        return (a * d - b * c) / np.sqrt(margins) if margins > 0 else NAN
    return np.sqrt(pearson(counts)[0] / counts.sum())


def cramers_v(counts):
    smaller = min(counts.shape) - 1
    return np.sqrt(pearson(counts)[0] / (counts.sum() * smaller)) if smaller > 0 else NAN


def contingency_coefficient(counts):
    chi2 = pearson(counts)[0]
    return np.sqrt(chi2 / (chi2 + counts.sum()))


def goodman_kruskal_lambda(counts):
    """(row dependent, column dependent, symmetric): the share of modal-guess errors
    removed by knowing the other variable."""
    n = counts.sum()
    rows, cols = counts.sum(axis=1), counts.sum(axis=0)
    best_given_column, best_given_row = counts.max(axis=0).sum(), counts.max(axis=1).sum()
    row_dependent = (best_given_column - rows.max()) / (n - rows.max()) if n > rows.max() else NAN
    col_dependent = (best_given_row - cols.max()) / (n - cols.max()) if n > cols.max() else NAN
    denominator = 2 * n - rows.max() - cols.max()
    symmetric = (_ratio(best_given_column + best_given_row - rows.max() - cols.max(), denominator))
    return row_dependent, col_dependent, symmetric


def _gini(proportions):
    return (1 - (proportions**2).sum()) / 2


def goodman_kruskal_tau(counts):
    """(column dependent, row dependent): the share of the qualitative (Gini) variance
    of one variable explained by the other."""
    n = counts.sum()
    rows, cols = counts.sum(axis=1), counts.sum(axis=0)
    if n <= 0 or (rows <= 0).any() or (cols <= 0).any():
        return NAN, NAN
    out = []
    for profiles, marginal, weights in (
        (counts / rows[:, None], cols / n, rows / n),       # column given each row
        ((counts / cols[None, :]).T, rows / n, cols / n),   # row given each column
    ):
        overall = _gini(marginal)
        within = sum(w * _gini(p) for w, p in zip(weights, profiles))
        out.append((overall - within) / overall if overall else NAN)
    return out[0], out[1]


def _entropy(counts, n):
    p = counts[counts > 0] / n
    return -(p * np.log(p)).sum()


def uncertainty_coefficient(counts):
    """(row dependent, column dependent, symmetric): Theil's U, mutual information over
    the entropy of the dependent variable."""
    n = counts.sum()
    h_row, h_col = _entropy(counts.sum(axis=1), n), _entropy(counts.sum(axis=0), n)
    mutual = h_row + h_col - _entropy(counts.ravel(), n)
    return (_ratio(mutual, h_row), _ratio(mutual, h_col), _ratio(2 * mutual, h_row + h_col))


# ---- ordinal ------------------------------------------------------------------------

def pairs(counts):
    """(concordant P, discordant Q, sum of squared row totals, of column totals, n).

    A case's concordant partners lie below and to the right of its cell, its discordant
    ones below and to the left; the categories are read in the table's order.
    """
    rows, cols = counts.shape
    concordant = discordant = 0.0
    for i in range(rows):
        for j in range(cols):
            concordant += counts[i, j] * counts[i + 1:, j + 1:].sum()
            discordant += counts[i, j] * counts[i + 1:, :j].sum()
    n = counts.sum()
    return (concordant, discordant, (counts.sum(axis=1) ** 2).sum(),
            (counts.sum(axis=0) ** 2).sum(), n)


def gamma(counts):
    p, q, _, _, _ = pairs(counts)
    return _ratio(p - q, p + q)


def tau_b(counts):
    p, q, tied_rows, tied_cols, n = pairs(counts)
    return _ratio(2 * (p - q), np.sqrt((n**2 - tied_rows) * (n**2 - tied_cols)))


def tau_c(counts):
    p, q, _, _, n = pairs(counts)
    m = min(counts.shape)
    return 2 * m * (p - q) / (n**2 * (m - 1)) if m >= 2 else NAN


def somers_d(counts):
    """(column dependent, row dependent, symmetric)."""
    p, q, tied_rows, tied_cols, n = pairs(counts)
    column_dependent, row_dependent = n**2 - tied_rows, n**2 - tied_cols
    return (_ratio(2 * (p - q), column_dependent), _ratio(2 * (p - q), row_dependent),
            _ratio(2 * (p - q), (column_dependent + row_dependent) / 2))


def cochran_armitage(counts):
    """(z, two-sided P, one-sided P) for a trend down the rows in the share falling in the
    first column. The rows are scored 1, 2, 3, ... in table order."""
    sizes, responses, n = counts.sum(axis=1), counts[:, 0], counts.sum()
    if n <= 0 or (sizes == 0).any():
        return NAN, NAN, NAN
    share = responses.sum() / n
    scores = np.arange(1, counts.shape[0] + 1, dtype=float)
    numerator = (scores * (responses - sizes * share)).sum()
    variance = share * (1 - share) * ((scores**2 * sizes).sum() - (scores * sizes).sum() ** 2 / n)
    if variance <= 0:
        return NAN, NAN, NAN
    z = numerator / np.sqrt(variance)
    p_two = 2 * stats.norm.sf(abs(z))
    return z, p_two, p_two / 2


def cumulative_chi_square(counts):
    """Hirotsu's cumulative chi-square: (statistic, effective df, P).

    The statistic adds the chi-squares of the r - 1 tables made by cutting the rows in
    two at each place. Those are correlated, so P comes from d * chi2(f) with d and f
    matched to the statistic's mean and variance under independence.
    """
    rows, cols = counts.shape
    n = counts.sum()
    if rows < 2 or cols < 2 or n <= 0:
        return NAN, NAN, NAN
    above = np.cumsum(counts, axis=0)[:-1]
    statistic = sum(pearson(np.vstack([above[i], counts.sum(axis=0) - above[i]]))[0]
                    for i in range(rows - 1))
    sizes = np.cumsum(counts.sum(axis=1))[:-1]
    sizes = sizes[(sizes > 0) & (sizes < n)]
    if sizes.size == 0:
        return NAN, NAN, NAN
    left, right = np.meshgrid(sizes, sizes, indexing="ij")
    lower, upper = np.minimum(left, right), np.maximum(left, right)
    rho2 = lower * (n - upper) / (upper * (n - lower))
    mean = (cols - 1) * sizes.size
    variance = 2 * (cols - 1) * rho2.sum()
    if mean <= 0 or variance <= 0:
        return NAN, NAN, NAN
    scale, df = variance / (2 * mean), 2 * mean**2 / variance
    return statistic, df, stats.chi2.sf(statistic / scale, df)


# ---- 2x2 ----------------------------------------------------------------------------

def fourfold(counts, confidence):
    """The 2x2 tab's measures by the app's names: (value, lower, upper), NaN where there
    is no interval.

    The first row is the positive test / exposed group, the first column the event /
    disease. Sensitivity and specificity read down the columns, the risks across the rows.
    """
    (a, b), (c, d) = counts
    n = a + b + c + d
    sensitivity, specificity = _ratio(a, a + c), _ratio(d, b + d)
    accuracy = _ratio(a + d, n)
    risk1, risk2 = _ratio(a, a + b), _ratio(c, c + d)
    difference = risk1 - risk2
    if not np.isfinite(difference):
        number_needed = NAN
    else:
        number_needed = float("inf") if difference == 0 else 1 / abs(difference)
    if min(a, b, c, d) > 0:
        # Ratios are skewed, so their limits are found on the log scale and mapped back.
        z = stats.norm.ppf(0.5 + confidence / 2)
        risk_ratio, risk_se = risk1 / risk2, np.sqrt(1 / a - 1 / (a + b) + 1 / c - 1 / (c + d))
        odds_ratio, odds_se = a * d / (b * c), np.sqrt(1 / a + 1 / b + 1 / c + 1 / d)
        ratios = {
            "Risk ratio": (risk_ratio, risk_ratio * np.exp(-z * risk_se), risk_ratio * np.exp(z * risk_se)),
            "Odds ratio": (odds_ratio, odds_ratio * np.exp(-z * odds_se), odds_ratio * np.exp(z * odds_se)),
        }
    else:
        # An empty cell leaves both ratios undefined on the log scale: no estimate is given.
        # In its place, Haldane's correction (0.5 added to every cell) and the conditional
        # maximum-likelihood odds ratio with Cornfield's exact interval.
        ratios = {"Risk ratio": (NAN, NAN, NAN), "Odds ratio": (NAN, NAN, NAN)}
        z = stats.norm.ppf(0.5 + confidence / 2)
        ha, hb, hc, hd = a + 0.5, b + 0.5, c + 0.5, d + 0.5
        risk_ratio = (ha / (ha + hb)) / (hc / (hc + hd))
        risk_se = np.sqrt(1 / ha - 1 / (ha + hb) + 1 / hc - 1 / (hc + hd))
        odds_ratio, odds_se = ha * hd / (hb * hc), np.sqrt(1 / ha + 1 / hb + 1 / hc + 1 / hd)
        conditional = contingency.odds_ratio(np.asarray(counts).astype(int), kind="conditional")
        interval = conditional.confidence_interval(confidence_level=confidence)
        ratios.update({
            "Risk ratio (Haldane correction)": (risk_ratio, risk_ratio * np.exp(-z * risk_se),
                                     risk_ratio * np.exp(z * risk_se)),
            "Odds ratio (Haldane correction)": (odds_ratio, odds_ratio * np.exp(-z * odds_se),
                                     odds_ratio * np.exp(z * odds_se)),
            "Odds ratio (conditional MLE)": (conditional.statistic, interval.low, interval.high),
        })
    single = {
        "Accuracy": accuracy,
        "Misclassification rate": 1 - accuracy,
        "Sensitivity": sensitivity,
        "Specificity": specificity,
        "NNT and NNH": number_needed,
        "Positive predictive value": _ratio(a, a + b),
        "Negative predictive value": _ratio(d, c + d),
        "Absolute risk reduction": difference,
        "Relative risk reduction": _ratio(difference, risk1),
        "Positive likelihood ratio": _ratio(sensitivity, 1 - specificity),
        "Negative likelihood ratio": _ratio(1 - sensitivity, specificity),
        "Fourfold point correlation": _ratio(a * d - b * c, np.sqrt((a + b) * (c + d) * (a + c) * (b + d))),
        "Yule's Q": _ratio(a * d - b * c, a * d + b * c),
    }
    return {**{name: (value, NAN, NAN) for name, value in single.items()}, **ratios}
```

## Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats
from cross_tabulation_measures import (
    cochran_armitage, contingency_coefficient, cramers_v, cumulative_chi_square, fourfold,
    gamma, goodman_kruskal_lambda, goodman_kruskal_tau, likelihood_ratio, mantel_haenszel,
    pearson, phi, somers_d, tau_b, tau_c, uncertainty_coefficient, yates,
)

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 30)
np.seterr(divide="ignore", invalid="ignore")  # 0/0 in an empty row of a stratum is NaN
NAN = float("nan")


def label(value):
    """A category's name as the app prints it: 2.0 read from a number column is "2"."""
    if isinstance(value, float) and value.is_integer():
        return str(int(value))
    return str(value)


def cp932(text):
    """Sort key for text: its Shift-JIS bytes, which put Japanese text in the order a
    Japanese Excel sorts it. For English text it is plain ASCII order."""
    try:
        return text.encode("cp932")
    except UnicodeEncodeError:
        return text.encode("utf-8")


def sorted_levels(values):
    """Categories in ascending order: numbers by value, text by cp932()."""
    levels = list(pd.unique(values))
    try:
        return sorted(levels, key=float)
    except (TypeError, ValueError):
        return sorted(levels, key=lambda level: cp932(str(level)))


def show(title, frame):
    print(f"\n{title}")
    print(frame)
    return frame
```

## The data

```python
# One row per case. Cases with a blank in any of the columns used are left out.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
row, column = "Plan", "Satisfaction"
used = [row, column]
frame = data[used].dropna().copy()
```

## Category order

```python
# The order matters beyond looks: every ordinal statistic reads the categories in it.
row_levels = sorted_levels(frame[row])
column_levels = sorted_levels(frame[column])


def tabulate(cases):
    """Counts of every row x column category pair among these cases, in the chosen order."""
    table = pd.crosstab(cases[row], cases[column])
    return table.reindex(index=row_levels, columns=column_levels).fillna(0)


table = tabulate(frame)
counts = table.to_numpy(float)
row_labels = [label(level) for level in table.index]
column_labels = [label(level) for level in table.columns]


def layout(matrix):
    """The cross tabulation as the app lays it out, margins included."""
    lines, names = [], []
    column_totals, n = matrix.sum(axis=0), matrix.sum()
    for name, body in [*zip(row_labels, matrix), ("Total", column_totals)]:
        total = body.sum()
        lines.append([*body, total])
        names.append(name)
        # Row percentages: each cell over its row's total, so every row sums to 1.
        lines.append([*(body / total), 1.0])
        names.append(f"{name} (share)")
    frame = pd.DataFrame(lines, index=names, columns=[*column_labels, "Total"])
    return frame
```

## Residual analysis

```python
# Under independence a cell expects (row total x column total) / n. The adjusted
# standardised residual divides observed - expected by its own standard error,
#   sqrt(expected x (1 - row total / n) x (1 - column total / n)),
# so it is close to standard normal: beyond +-1.96 a cell is significant at 5%.
n = counts.sum()
expected = np.outer(counts.sum(axis=1), counts.sum(axis=0)) / n
residual = counts - expected
adjusted = residual / np.sqrt(expected * np.outer(1 - counts.sum(axis=1) / n, 1 - counts.sum(axis=0) / n))
p_two = 2 * stats.norm.sf(np.abs(adjusted))  # sf is 1 - Phi without losing digits
if (expected < 5).any():
    print(f"\n{(expected < 5).sum()} cells expect fewer than 5: the chi-square approximation is rough.")




def matrix_table(title, matrix):
    return show(title, pd.DataFrame(matrix, index=row_labels, columns=column_labels))


expected_table = matrix_table("Expected frequency", expected)
residual_table = matrix_table("Residual", residual)
adjusted_table = matrix_table("Adjusted standardised residual", adjusted)
p_two_table = matrix_table("Adjusted standardised residual: two-sided P", p_two)
p_one_table = matrix_table("Adjusted standardised residual: one-sided P", p_two / 2)
```

## Statistics: the general, ordinal, 2×2 and other statistics

```python
# Each ticked statistic is stored under its name in the app, and each tab's table lists
# them in the order they were ticked. With a layer, every tab is computed for each
# stratum's table and then for the whole table. A stratum's table is its own: a row or
# column with no case in that stratum is left out, as pd.crosstab of the stratum's cases
# leaves it out, so it adds nothing to the degrees of freedom and takes no score. A
# statistic the table cannot carry -- a 2x2 measure on a larger table, an ordinal measure
# on a nominal axis -- is a row of NaN, where the app prints its reason.
chi_square_statistics = ["Pearson's chi-square", "Residual analysis", "Likelihood ratio"]
nominal_statistics = ["Phi coefficient", "Cramer's V", "Contingency coefficient", "Goodman-Kruskal's lambda", "Goodman-Kruskal's tau", "Uncertainty coefficient"]
ordinal_statistics = []
two_by_two_statistics = []
other_statistics = []
confidence = 95 / 100
NONE = (NAN, NAN, NAN)
# The ordinal measures need both axes ordinal or interval-ratio.
ordinal_axes = "Nominal" not in ("Nominal", "Nominal")


def collect(tables, title, found, ticked, columns):
    """Add one tab's table: the rows of its ticked statistics, in the order ticked."""
    rows = {label: value for name in ticked if name in found for label, value in found[name].items()}
    if rows:
        tables.append(show(title, pd.DataFrame(list(rows.values()), index=list(rows), columns=columns)))


tests, nominals, ordinals, fourfolds, others = [], [], [], [], []
scopes = []  # (title, table, is it one stratum's table?)
scopes.append(("whole table", counts, False))
for scope, matrix, per_stratum in scopes:
    found = {}  # name in the app -> {row label: numbers}
    fourfold_table = matrix.shape == (2, 2)
    found["Pearson's chi-square"] = {"Pearson": pearson(matrix)}
    found["Likelihood ratio"] = {"Likelihood ratio (G2)": likelihood_ratio(matrix)}
    found["Phi coefficient"] = {"Phi": phi(matrix)}
    found["Cramer's V"] = {"Cramer's V": cramers_v(matrix)}
    found["Contingency coefficient"] = {"Contingency coefficient": contingency_coefficient(matrix)}
    row_dependent, column_dependent, symmetric = goodman_kruskal_lambda(matrix)
    found["Goodman-Kruskal's lambda"] = {"lambda (symmetric)": symmetric, "lambda (row dependent)": row_dependent,
                                         "lambda (column dependent)": column_dependent}
    column_dependent, row_dependent = goodman_kruskal_tau(matrix)
    found["Goodman-Kruskal's tau"] = {"tau (column dependent)": column_dependent, "tau (row dependent)": row_dependent}
    row_dependent, column_dependent, symmetric = uncertainty_coefficient(matrix)
    found["Uncertainty coefficient"] = {"uncertainty (symmetric)": symmetric, "uncertainty (row dependent)": row_dependent,
                        "uncertainty (column dependent)": column_dependent}
    if two_by_two_statistics and fourfold_table:
        # The 2x2 measures (value, lower, upper): the first row is the positive test /
        # exposed group, the first column the event.
        measures = fourfold(matrix, confidence)
        found.update({name: {name: measures[name]} for name in two_by_two_statistics if name in measures})
        # With an empty cell the two ratios are blank, and the corrected rows follow them.
        for name, corrected in (("Risk ratio", ["Risk ratio (Haldane correction)"]),
                                ("Odds ratio", ["Odds ratio (Haldane correction)", "Odds ratio (conditional MLE)"])):
            if name in found:
                found[name].update({label: measures[label] for label in corrected if label in measures})
    elif two_by_two_statistics:
        found[two_by_two_statistics[0]] = {"2x2 statistics": NONE}  # the table is not 2x2
    collect(tests, f"Chi-square tests ({scope})", found, chi_square_statistics, ["chi-square", "df", "P"])
    collect(nominals, f"Nominal measures ({scope})", found, nominal_statistics, ["value"])
    collect(ordinals, f"Ordinal measures ({scope})", found, ordinal_statistics, ["value", "df", "P"])
    collect(fourfolds, f"2x2 measures ({scope}, {confidence:.0%} interval)", found, two_by_two_statistics,
            ["value", "lower", "upper"])
    collect(others, f"Other ({scope})", found, other_statistics, ["value", "df", "P"])
```

## What the result says

This run adds a second test and measures of strength. The likelihood-ratio chi-square (81.67) agrees with Pearson's (79.55): both P values are far below 0.01, so the link between plan and satisfaction is real.

A test does not say how strong a link is. Cramer's V, which runs from 0 (no link) to 1 (perfect link), is 0.25, and the contingency coefficient is 0.34. The other measures ask how much knowing the plan helps to guess the satisfaction score. Goodman-Kruskal's lambda is 0.044 and tau is 0.028 with satisfaction as the dependent variable, and the uncertainty coefficient is 0.043. All three are close to 0.

So the link is clear but only moderate. The plan alone is a weak guide to how satisfied one subscriber is.
