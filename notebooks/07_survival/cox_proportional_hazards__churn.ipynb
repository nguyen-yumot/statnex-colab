These cells save `cox_diagnostics.py`: a step that no library does, written out. The analysis imports it.

```python
%%writefile cox_diagnostics.py
"""Two checks on a fitted Cox model that statsmodels does not make: does each hazard
ratio hold still over time, and how well does the model order who fails first.

proportional_hazards_test is Grambsch and Therneau's (1994) score test, as R's
survival::cox.zph computes it since survival 3.0, on its default time scale
(transform = "km"): each coefficient may drift as b + theta g(t), g(t) being one minus
the Kaplan-Meier estimate just before t, and theta = 0 is tested at the fitted b. Breslow's ties, as the model is fitted.

harrell_c is Harrell's concordance with the standard error R's survival::concordance
reports, its infinitesimal jackknife.
"""

import numpy as np
import pandas as pd
from scipy import stats


def proportional_hazards_test(x, times, events, beta, names):
    """chi-square, df and P per covariate, and for all of them together (GLOBAL)."""
    x = x - x.mean(axis=0)  # the risk-set covariance is unchanged; its digits are safer
    p = x.shape[1]
    risk = np.exp(x @ beta)
    # R's default time scale, g(t) = 1 - S(t-): one minus the Kaplan-Meier estimate just
    # before each event time, from all cases pooled (cox.zph's transform = "km").
    event_times, deaths = np.unique(times[events == 1], return_counts=True)
    at_risk_n = np.array([(times >= t).sum() for t in event_times])
    km_before = np.r_[1.0, np.cumprod(1 - deaths / at_risk_n)[:-1]]
    scale = dict(zip(event_times, 1 - km_before))
    mean_scale = np.average(1 - km_before, weights=deaths)
    score = np.zeros(p)
    information = np.zeros((2 * p, 2 * p))
    for t in np.unique(times[events == 1]):
        at_risk = times >= t
        dying = (times == t) & (events == 1)
        w = risk[at_risk] / risk[at_risk].sum()
        mean = w @ x[at_risk]
        centred = x[at_risk] - mean
        v = centred.T @ (centred * w[:, None])
        g = scale[t] - mean_scale
        # The Schoenfeld residuals of the deaths at t, weighted by g(t), score theta.
        score += g * (x[dying] - mean).sum(axis=0)
        # Information for (b, theta): the risk-set covariance times 1, g and g^2.
        information += dying.sum() * np.kron([[1, g], [g, g * g]], v)
    rows = []
    for j in range(p):
        keep = [*range(p), p + j]
        u = np.r_[np.zeros(p), score[j]]
        rows.append([u @ np.linalg.solve(information[np.ix_(keep, keep)], u), 1])
    u = np.r_[np.zeros(p), score]
    rows.append([u @ np.linalg.solve(information, u), p])
    table = pd.DataFrame(rows, index=[*names, "GLOBAL"], columns=["chi-square", "df"])
    table["P"] = stats.chi2.sf(table["chi-square"], table["df"])
    return table


def harrell_c(times, events, risk):
    """(C, SE). A pair is comparable when the shorter time is an event -- one censored at
    that very time still counts as at risk; two events at one time are not compared.
    It is concordant when the case that failed first has the higher risk score, and a
    tied score counts a half."""
    n = len(times)
    counts = np.zeros((n, 3))  # each case's concordant, discordant and tied pairs
    for i in np.flatnonzero(events == 1):
        later = (times > times[i]) | ((times == times[i]) & (events == 0))
        concordant = later & (risk < risk[i])
        discordant = later & (risk > risk[i])
        tied = later & (risk == risk[i])
        counts[i] += [concordant.sum(), discordant.sum(), tied.sum()]
        counts[:, 0] += concordant
        counts[:, 1] += discordant
        counts[:, 2] += tied
    # Every pair was counted at both its ends.
    concordant, discordant, tied = counts.sum(axis=0) / 2
    comparable = concordant + discordant + tied
    somers_d = (concordant - discordant) / comparable
    # The infinitesimal jackknife: each case's influence on C, from the pairs it is in.
    influence = (counts[:, 0] - counts[:, 1] - counts.sum(axis=1) * somers_d) / (2 * comparable)
    return (somers_d + 1) / 2, np.sqrt((influence ** 2).sum())
```

## Covariates that are linear combinations of the others

```python
# A covariate whose removal leaves the design's rank unchanged is a combination of the
# rest; the coefficients would not be unique, so it has to be dropped first.
design = np.column_stack([np.ones(n), frame[covariates].to_numpy(float)])
rank = np.linalg.matrix_rank(design)
dependent = [name for i, name in enumerate(covariates) if rank < design.shape[1]
             and np.linalg.matrix_rank(np.delete(design, i + 1, axis=1)) == rank]
print("\nLinear combinations:", ", ".join(dependent) if dependent else "none")
```

## The Cox model

```python
# The model is h(t | x) = h0(t) exp(b1 x1 + ... + bp xp): every case shares one baseline
# hazard h0(t), the momentary rate of the event, and its covariates multiply it. exp(b)
# is the hazard ratio, the factor by which one more unit of a covariate multiplies that
# rate at every moment, the other covariates held fixed. The model assumes the factor
# is the same at every time (proportional hazards, tested below) and that censoring is
# unrelated to the risk of the event.
# statsmodels' PHReg with Breslow's handling of tied times: every death at a tied time
# is scored against the same full risk set, as SAS and Stata do by default (R's coxph
# defaults to Efron's). Each fit is kept by its set of covariates.
# PHReg is given each covariate minus its mean, as R's coxph centres them. The
# coefficients, the log likelihood and its derivatives are the same either way, but a
# covariate far from 0 (a date serial, 45000) would lose digits to rounding.
centred = frame[covariates] - frame[covariates].mean()
fits = {}


def fit(active):
    """(coefficients, log likelihood, information matrix) of the model on ``active``."""
    key = tuple(active)
    if key not in fits:
        if active:
            model = PHReg(times, centred[active].to_numpy(float), status=events, ties="breslow")
            result = model.fit()
            fits[key] = (result.params, result.llf, -model.hessian(result.params))
        else:  # no covariates: the log likelihood at beta = 0
            model = PHReg(times, centred[covariates[:1]].to_numpy(float), status=events,
                          ties="breslow")
            fits[key] = (np.zeros(0), model.loglike(np.zeros(1)), np.zeros((0, 0)))
    return fits[key]


def score_test(columns, beta):
    """Rao's score test of the coefficients that are 0 in ``beta``: how steeply the log
    likelihood climbs from there, judged by its variance. No bigger model is fitted, which
    is why it is the test for letting a covariate in."""
    model = PHReg(times, centred[columns].to_numpy(float), status=events, ties="breslow")
    u = model.score(beta)
    inverse = np.linalg.inv(-model.hessian(beta))
    free = beta == 0
    statistic = u[free] @ inverse[np.ix_(free, free)] @ u[free]
    return statistic, int(free.sum()), stats.chi2.sf(statistic, free.sum())


def outside_tests(active):
    """Score test of each covariate not in the model, entered alone at the fitted beta."""
    beta = fit(active)[0]
    return [(name, *score_test([*active, name], np.append(beta, 0.0)))
            for name in covariates if name not in active]


def wald(active):
    """(coefficient, SE, Wald chi-square, P) of each covariate in the model."""
    beta, _, information = fit(active)
    se = np.sqrt(np.diag(np.linalg.inv(information)))
    chi2 = (beta / se) ** 2
    return beta, se, chi2, stats.chi2.sf(chi2, 1)
```

## Variable selection

```python
enter_p, remove_p = 0.2, 0.2
criteria = [enter_p, remove_p]
active = []
steps = [{"active": list(active), "entered": "", "removed": ""}]
# Entry uses the score test and removal the Wald test; near a threshold they can disagree
# and pass a covariate in and out forever. A move back to a model already visited ends
# the search.
# The P values printed for the chosen model do not allow for the search that chose it:
# they come out smaller, and the hazard ratios further from 1, than a test planned in
# advance would give. Treat stepwise selection as a screen, not as proof.
seen = {frozenset(active)}
for number in range(1, 2 * len(covariates) + 2):
    entered = removed = ""
    before = list(active)
    # Forward: of the covariates outside, the one with the smallest score-test P enters,
    # if that P is below the entry criterion.
    candidates = outside_tests(active)
    if candidates:
        best = min(candidates, key=lambda c: c[3])
        if best[3] < enter_p:
            active = [*active, best[0]]
            entered = best[0]
    # Backward: if nothing entered, the covariate with the largest Wald P leaves, if that
    # P is above the removal criterion.
    if not entered and active:
        p_values = wald(active)[3]
        worst = int(np.argmax(p_values))
        if p_values[worst] > remove_p:
            removed = active[worst]
            active = [name for i, name in enumerate(active) if i != worst]
    if frozenset(active) in seen:
        active = before
        break  # nothing moved, or back where it has been: the search has settled
    seen.add(frozenset(active))
    steps.append({"active": list(active), "entered": entered, "removed": removed})
# Each row of the process table announces the move that follows it.
for this, following in zip(steps, steps[1:] + [None]):
    this["next"] = ((following["entered"], following["removed"]) if following
                    else ("(none)", ""))
```

## The variable selection process

```python
# For each step: -2 log likelihood and AIC; the score test of the whole model against
# none; and the likelihood-ratio change from the step before.
null_loglik = fit([])[1]
previous_loglik, previous_size = null_loglik, 0
rows = []
for number, step in enumerate(steps):
    active = step["active"]
    loglik = fit(active)[1]
    deviance = -2 * loglik
    if active:
        score, df, score_p = score_test(active, np.zeros(len(active)))
        change = 2 * (loglik - previous_loglik)
        change_df = abs(len(active) - previous_size) or len(active)
        change_p = stats.chi2.sf(abs(change), change_df)
        rows.append([deviance, deviance + 2 * len(active), score, df, score_p,
                     abs(change), change_df, change_p])
    else:
        rows.append([deviance, deviance, np.nan, np.nan, np.nan, np.nan, np.nan, np.nan])
    previous_loglik, previous_size = loglik, len(active)
process = pd.DataFrame(rows, index=[f"step {i}" for i in range(len(steps))],
                       columns=["-2LL", "AIC", "score chi2", "df", "P", "change chi2",
                                "change df", "change P"])
process["entered"] = [step["next"][0] for step in steps]
process["removed"] = [step["next"][1] for step in steps]
print()
print(process)
```

## Covariates in and out of the model, step by step

```python
# In: Wald tests and hazard ratios exp(b) with their interval, in the order the
# covariates were given. Out: the score test for entering each one.
inside_rows, outside_rows = [], []
for number, step in enumerate(steps):
    active = step["active"]
    if not active:
        inside_rows.append([number, "(none)"] + [np.nan] * 8)
    else:
        beta, se, chi2, p = wald(active)
        for name in [c for c in covariates if c in active]:
            i = active.index(name)
            inside_rows.append([number, name, beta[i], se[i], chi2[i], 1, p[i], np.exp(beta[i]),
                                np.exp(beta[i] - z * se[i]), np.exp(beta[i] + z * se[i])])
    tests = outside_tests(active)
    if not tests:
        outside_rows.append([number, "(none)", np.nan, np.nan, np.nan])
    for name, statistic, df, p in tests:
        outside_rows.append([number, name, statistic, df, p])
inside = pd.DataFrame(inside_rows, columns=["step", "covariate", "b", "SE", "chi2", "df", "P",
                                            "exp(b)", "lower", "upper"])
outside = pd.DataFrame(outside_rows, columns=["step", "covariate", "chi2", "df", "P"])
print()
print(inside.to_string(index=False))
print()
print(outside.to_string(index=False))
```

## The final model

```python
# The last step's model. Coefficients are listed in the order the covariates entered.
final = steps[-1]["active"]
result = process.iloc[-1, :8].to_numpy(float)
print()
print(pd.Series(result, index=process.columns[:8]))
if final:
    beta, se, chi2, p = wald(final)
    coefficients = pd.DataFrame({"b": beta, "SE": se, "chi2": chi2, "df": 1, "P": p,
                                 "exp(b)": np.exp(beta), "lower": np.exp(beta - z * se),
                                 "upper": np.exp(beta + z * se)}, index=final)
    print()
    print(coefficients)

    # === Simulation ===
    # The hazard ratio of a case with every covariate at its smallest value (A) and at its
    # largest (B), against a case with every covariate at 0. When the covariates lie far
    # from 0 (a date serial such as 45000) that ratio is larger than the largest number a
    # computer holds, about 1e308, and exp() gives inf: it is left blank (NaN). The ratio
    # of B to A, exp(b (B - A)), is still a number.
    low, high = frame[final].min().to_numpy(), frame[final].max().to_numpy()
    simulation = pd.DataFrame({"b": beta, "A": low, "B": high}, index=final)
    with np.errstate(over="ignore"):
        ratios = np.exp([beta @ low, beta @ high])
    simulation.loc["hazard ratio"] = [np.nan, *np.where(np.isinf(ratios), np.nan, ratios)]
    print()
    print(simulation)

    # === Statistics by case ===
    # Breslow's cumulative hazard adds, at each event time, the deaths over the sum of the
    # risks exp(xb) of the cases still at risk. It is summed here because PHReg's
    # baseline_cumulative_hazard reports the value just before each event time, without
    # that time's own step. Each risk is taken against the mean case, exp((x - mean) b),
    # and divided by the largest, so no exp() can overflow or round to 0 when xb is large
    # (a date serial as a covariate, say). A case's cumulative hazard H(t) is that sum
    # times its own risk, and its survival is S(t) = exp(-H(t)): a higher-risk case's
    # curve falls faster. The baseline H0(t), every covariate at 0, is the same sum moved
    # back to that case; a value larger than a computer number holds is left blank (NaN).
    x = frame[final].to_numpy(float)
    relative = (x - x.mean(axis=0)) @ beta
    top = relative.max()
    risk = np.exp(relative - top)  # each between 0 and 1
    grid = np.unique(times[events == 1])
    steps_h = [events[times == t].sum() / risk[times >= t].sum() for t in grid]
    cumulative = np.cumsum(steps_h)
    position = np.searchsorted(grid, times, side="right") - 1
    summed = np.where(position >= 0, cumulative[np.maximum(position, 0)], 0.0)
    hazard = summed * risk
    survival = np.exp(-hazard)
    with np.errstate(over="ignore", invalid="ignore", divide="ignore"):
        to_zero = np.exp(-(top + x.mean(axis=0) @ beta))  # the mean case to every covariate at 0
        baseline_hazard = np.where(summed > 0, summed * to_zero, 0.0)
        log_log = np.log(np.where(hazard > 0, hazard, np.nan))
    baseline_survival = np.exp(-baseline_hazard)
    baseline_hazard = np.where(np.isinf(baseline_hazard), np.nan, baseline_hazard)
    by_case = pd.DataFrame({"time": times, "status": status_values, "S0": baseline_survival,
                            "H0": baseline_hazard, "S": survival, "H": hazard,
                            "log-log S": log_log}, index=np.arange(1, n + 1))
    print()
    print(by_case)

    # === Proportional hazards, and Harrell's C ===
    # The score test on the Schoenfeld residuals with g(t) = 1 - S(t-), one minus the
    # Kaplan-Meier estimate, as R's cox.zph() by default; Harrell's C with the infinitesimal-jackknife
    # SE of R's concordance(), and the interval C ± z SE (cox_diagnostics.py). Neither is in
    # statsmodels; lifelines' proportional_hazard_test is the older approximation (Stata's
    # -estat phtest-), so the exact test is written out.
    # How to read them: a small P for a covariate says its hazard ratio changes over time,
    # so its single exp(b) is an average over the follow-up; a large P is no proof that it
    # is constant, only no evidence against it. Harrell's C takes every pair of cases in
    # which one had the event first and counts the share the model gave the higher risk
    # (0.5 is a coin toss, 1 a perfect ordering): how well the model ranks cases, which a
    # significant hazard ratio does not promise.
    x_final = frame[final].to_numpy(float)
    ph_test = proportional_hazards_test(x_final, times, events, beta, final)
    c_index, c_se = harrell_c(times, events, x_final @ beta)
    concordance = [c_index, c_se, c_index - z * c_se, c_index + z * c_se]
    print()
    print(ph_test)
    print(f"Harrell's C {c_index:.4f} (SE {c_se:.4f}), interval {concordance[2]:.4f} to "
          f"{concordance[3]:.4f}")
```

## What the result says

The Cox model relates customer features to the hazard, the rate at which customers cancel at each moment. Forward stepwise selection adds one feature at a time, while it improves the model. It added the promo, then the monthly price. The signup channel stayed out (P = 0.350).

Exp(coefficient) is the hazard ratio: how many times the cancelling rate is multiplied. For promo customers it is 2.60, with a confidence interval (the range of values that fit the data) from 1.59 to 4.24. Each extra dollar of monthly price multiplies the rate by 1.078 (P = 0.008).

The test of proportional hazards gives P = 0.029 for the promo, so its effect may change over time, and 2.60 is an average. Customers chose how to join, so this does not prove that the promo drives them away.
