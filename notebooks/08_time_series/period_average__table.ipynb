**The story.** A bubble-tea chain has its quarterly revenue from 2020 to 2024, in thousands of dollars. The table is laid out the way a finance team often keeps it in a spreadsheet: one row per quarter, one column per year.

**The question.** Can the seasonal pattern be measured straight from a table with one column per year?

## Libraries

```python
import numpy as np
import pandas as pd

pd.set_option("display.width", 120)
```

## The data

```python
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
# A ready-made table: one row per period (quarter, month...), one column per cycle
# (year).
cycles = ["2020", "2021", "2022", "2023", "2024"]
table = data[cycles].astype(float)
table.index = pd.RangeIndex(1, len(table) + 1, name="Period")
```

## Period averages and the seasonal index

```python
# Average each period across the cycles (skipping a missing one), then divide by the
# mean of those averages: an index above 1 marks a period above the yearly level, so
# 1.30 reads "30% above an average period". No library function computes this simple
# index, so it is written out; it is two lines of averages.
# The method assumes the series has no trend. In a growing series the later periods
# of every cycle sit higher on the trend, so their indices come out too high and the
# early periods' too low. Methods that take the trend out first (the link relative
# method, or the ratio to a centred moving average in statsmodels' seasonal_decompose)
# avoid that.
table["Period average"] = table[cycles].mean(axis=1)
table["Period index"] = table["Period average"] / table["Period average"].mean()
print(table)
```

## What the result says

Here the same revenue is entered as a table: one row for each of the 4 quarters and one column for each year from 2020 to 2024. The seasonal average method works the same way. It takes the mean of each row, the period average, and divides it by the mean of all quarters to get the period index, where 1 means an average quarter.

The period averages are 994.2, 1371, 1651.2 and 1056.4, in \$000. The period indices are 0.784, 1.081, 1.302 and 0.833. These are exactly the values the list layout gives: summer (the third quarter) almost a third above an average quarter, and the first quarter more than a fifth below.

So the layout does not matter. Only the arrangement of the numbers changed, not the calculation. Use the layout your data already has.
