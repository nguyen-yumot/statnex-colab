## Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.width", 160)
```

## The data

```python
# One column per variable, one row per case. Blank and text cells are missing.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
columns = ["Temperature (C)", "Visitors (thousands)", "Ice cream sales ($000)", "Sunscreen sales ($000)", "Hot coffee sales ($000)", "Rainy days"]
numeric = data[columns].apply(pd.to_numeric, errors="coerce")
# Listwise deletion: a case missing any of the variables is dropped from all of them.
numeric = numeric.dropna()
counts = pd.DataFrame(len(numeric), index=columns, columns=columns)
```

## Correlation matrix

```python
# Pearson's r for every pair: how closely the two variables follow a straight line,
# from -1 (falling line) through 0 (no straight-line link) to +1 (rising line). It
# measures association only, never which variable drives which.
corr = numeric.corr(method="pearson")
print(corr)
```

## Test of zero correlation

```python
# scipy.stats.pearsonr() for each pair: t = r √(n - 2) / √(1 - r²) on n - 2 df,
# two-sided. A small P says a straight-line link this strong would be rare if the
# true r were 0; it says nothing about how strong the link is. It assumes independent
# cases (months in a row often are not) and roughly normal scatter. Each pair uses
# the cases observed on both, which is pairwise deletion when the data still have
# gaps. The app prints the P value above the diagonal and * (P < 0.05) / ** (P < 0.01)
# below it.
r = corr.to_numpy()
n = counts.to_numpy(float)
pairs = [(i, j) for i in range(len(columns)) for j in range(i + 1, len(columns))]
tests = {}
p_values = np.full((len(columns), len(columns)), np.nan)
for i, j in pairs:
    if n[i, j] > 2 and abs(r[i, j]) < 1:
        both = numeric[[columns[i], columns[j]]].dropna()
        tests[i, j] = stats.pearsonr(both.iloc[:, 0], both.iloc[:, 1])
        p_values[i, j] = p_values[j, i] = tests[i, j].pvalue
print()
print(pd.DataFrame(p_values, index=columns, columns=columns))
# Every pair once, in the order of the matrix's upper triangle: the list `pairs`
# above.
```

## Partial correlation matrix

```python
# The correlation of each pair with all the other variables held fixed: the link
# left between two variables once the straight-line effect of every other variable
# is taken out of both. A large r with a small partial r points to a common driver.
# scipy and statsmodels have no partial-correlation matrix, so it is read off the
# inverse of the correlation matrix: -inverse[i, j] / √(inverse[i, i] inverse[j, j]).
inverse = np.linalg.inv(r)
scale = np.sqrt(np.diag(inverse))
partial = -inverse / np.outer(scale, scale)
np.fill_diagonal(partial, 1.0)
partial = pd.DataFrame(partial, index=columns, columns=columns)
print()
print(partial)
```

## Test of zero partial correlation

```python
# Each partial correlation holds the other k - 2 variables fixed, and each one held
# fixed costs a degree of freedom: df = n - 2 - (k - 2) = n - k, with n the cases
# observed on every variable. t = r √df / √(1 - r²), two-sided. With few cases and
# many variables, df is small and even a sizeable partial r can be non-significant:
# that means "too little data to tell", not "no link".
df_partial = len(numeric.dropna()) - len(columns)
rp = partial.to_numpy()
with np.errstate(divide="ignore", invalid="ignore"):
    partial_p = 2 * stats.t.sf(np.abs(rp) * np.sqrt(df_partial) / np.sqrt(1 - rp**2), df_partial)
partial_p[(np.abs(rp) >= 1) | (df_partial <= 0)] = np.nan
print(f"\ndf = {df_partial}")
print(pd.DataFrame(partial_p, index=columns, columns=columns))
```

## Confidence interval for the population correlation

```python
# pearsonr()'s confidence_interval(), built on Fisher's z: artanh r is close to normal
# with SE 1/√(n - 3), so the interval is built there and taken back with tanh -- which
# keeps it inside (-1, 1) and makes it lopsided near ±1. It assumes the pair is
# roughly bivariate normal. Read it as the range of true correlations the data fit;
# the 95% describes the method's hit rate over repeated samples, not this interval.
level = 95 / 100
ci_rows = []
for i, j in pairs:
    n_ij, r_ij = n[i, j], r[i, j]
    if n_ij > 3 and abs(r_ij) < 1:
        lower, upper = tests[i, j].confidence_interval(confidence_level=level)
    else:
        lower = upper = np.nan
    ci_rows.append([columns[i], columns[j], n_ij, r_ij, lower, upper])
ci_table = pd.DataFrame(ci_rows, columns=["variable 1", "variable 2", "n", "r", "lower", "upper"])
print()
print(ci_table.to_string(index=False))
```

## Test against a comparison value

```python
# Is the true correlation different from rho0 (not from 0)? pearsonr() only tests 0,
# so this is written out, on Fisher's z again: z = (artanh r - artanh rho0) √(n - 3),
# two-sided against the standard normal. The test is signed: a correlation of -0.95
# is far from +0.9, so read it for the pairs whose sign matches rho0.
rho0 = float(0.9)
compare_rows = []
for i, j in pairs:
    n_ij, r_ij = n[i, j], r[i, j]
    if n_ij > 3 and abs(r_ij) < 1:
        z = (np.arctanh(r_ij) - np.arctanh(rho0)) * np.sqrt(n_ij - 3)
        p = 2 * stats.norm.sf(abs(z))
    else:
        z = p = np.nan
    compare_rows.append([columns[i], columns[j], n_ij, r_ij, z, p])
compare_table = pd.DataFrame(compare_rows, columns=["variable 1", "variable 2", "n", "r", "z", "P"])
print()
print(compare_table.to_string(index=False))
```

## What the result says

This run adds a confidence interval and tests. A correlation shows how closely two measures move together. Between ice cream and sunscreen sales it is 0.96. Its 95% confidence interval, the range of values the data agree with for the true correlation, runs from 0.85 to 0.99.

The test against the comparison value 0.9 gives P = 0.175. A P value is the chance of a result this far from 0.9 if the true correlation were 0.9. It is not small, so the data do not show that the true correlation is above 0.9.

The partial correlation, the link after the other four measures are held constant, is -0.18 with P = 0.67. A link between the two products beyond the weather is not shown. That does not prove there is none.
