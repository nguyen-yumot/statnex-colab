## Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.width", 120)
```

## The data

```python
# One column names each case's stratum; the others are the variables to summarise. A
# row with a blank or non-numeric value in any of them is dropped, so every variable
# is summarised on the same cases.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
factor = "App"
columns = ["Delivery time (min)", "Distance (km)", "Order total ($)", "Tip ($)"]
frame = data[[factor, *columns]].copy()
frame[columns] = frame[columns].apply(pd.to_numeric, errors="coerce")
frame = frame.dropna()
```

## Descriptive statistics

```python
# All cases together. Variance and SD divide by n - 1.
overall = pd.DataFrame({
    name: {"n": frame[name].size, "mean": np.mean(frame[name]),
           "variance": np.var(frame[name], ddof=1), "sd": np.std(frame[name], ddof=1)}
    for name in columns
})
print(overall)
```

## Descriptive statistics by stratum

```python
# Strata are listed in the reverse of the order they first appear in the data.
grouped = frame.groupby(factor, sort=False)[columns]
by_n = grouped.count().iloc[::-1]
by_mean = grouped.mean().iloc[::-1]
by_sd = grouped.std(ddof=1).iloc[::-1]
print("\nn")
print(by_n)
print("\nmean")
print(by_mean)
print("\nstandard deviation")
print(by_sd)
```

## Correlation ratio with the stratifying factor

```python
# η² (eta squared) is the share of a variable's total variation that lies between the
# strata: SS between / (SS between + SS within). 0 means the strata do not differ at
# all; 1 means knowing the stratum tells you the value exactly. SciPy has no function
# for η², so the two sums of squares are written out. Its F test is the one-way
# ANOVA's, stats.f_oneway: a small P says the stratum means are not all equal. The F
# test assumes roughly normal values with similar spreads in every stratum; with
# equal stratum sizes it tolerates moderate differences in spread.
eta = {}
for name in columns:
    groups = [g.to_numpy(float) for _, g in frame.groupby(factor, sort=False)[name]]
    values = np.concatenate(groups)
    grand = values.mean()
    between = sum(g.size * (g.mean() - grand) ** 2 for g in groups)
    within = sum(((g - g.mean()) ** 2).sum() for g in groups)
    test = stats.f_oneway(*groups)
    eta[name] = {"eta squared": between / (between + within), "F": test.statistic,
                 "df1": len(groups) - 1, "df2": values.size - len(groups), "P": test.pvalue}
eta = pd.DataFrame(eta)
print()
print(eta)
```

## What the result says

The mean (average) of each measure differs clearly between the apps. Delivery time is 24.46 minutes for ZipEats, 32.94 for FeastNow and 39.76 for UrbanBite.

The correlation ratio η² is the share of a measure's variation that the app explains. It is 0.778 for order total and 0.766 for tip, over three quarters. It is 0.656 for distance and 0.575 for delivery time. Each F test compares the differences between the apps with the variation inside them. Its P value, the chance of differences this large if the apps were really alike, is practically zero. So these differences are far too large to be chance.

So the app explains most of the differences between orders. But UrbanBite also drives furthest, 4.86 km on average against 1.92 km for ZipEats. These tables cannot say whether the app itself is slow or the trips are just longer.
