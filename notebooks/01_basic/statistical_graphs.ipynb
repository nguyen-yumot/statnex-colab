# Three apps from one database table: seven views at once

**Statistical Graphs (Database Format)** · a worked example from the StatNex teaching guide

**The story.** An analyst has 150 food-delivery orders, 50 from each of three apps: ZipEats, FeastNow and UrbanBite. They are stored the way most order databases store them: one row per order, with an App column and the delivery time in minutes. One run gives seven different views of the data at once.

**The question.** How do the apps compare on delivery time across all the standard graphs, from a single table?

**To run it**, choose **Runtime → Run all**. The first cells save the table in this session. The rest is the calculation StatNex runs, written out in Python. It does not use StatNex, and every number it prints was checked against the app's.

## Libraries

```python
from decimal import Decimal
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.width", 120)
```

## The data

```python
# One column holds the values, another says which stratum each value belongs to. Rows
# with a blank in either, or a value that is not a number, are dropped. Strata are
# listed in the reverse of the order they first appear, and each stratum's values are
# sorted.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
factor = "App"
value = "Delivery time (min)"
frame = data[[factor, value]].copy()
frame[value] = pd.to_numeric(frame[value], errors="coerce")
frame = frame.dropna()
levels = list(dict.fromkeys(frame[factor]))[::-1]
series = {level: np.sort(frame.loc[frame[factor] == level, value].to_numpy(float))
          for level in levels}
```

## Descriptive statistics

```python
# Variance and SD divide by n - 1, SE = SD / √n, CV = SD / mean. Quartiles interpolate
# linearly between the sorted values (NumPy's default, Excel's QUARTILE.INC).
def describe(x):
    q1, median, q3 = np.percentile(x, [25, 50, 75])
    mean, sd = np.mean(x), np.std(x, ddof=1)
    return {"n": x.size, "mean": mean, "variance": np.var(x, ddof=1), "sd": sd,
            "se": stats.sem(x), "min": x.min(), "q1": q1, "median": median, "q3": q3,
            "max": x.max(), "iqr": q3 - q1, "cv": sd / mean if mean != 0 else np.nan}


summary = pd.DataFrame({level: describe(x) for level, x in series.items()})
print(summary)
```

## Frequency tables

```python
# Boundary i is start + i × width, rounded to the decimals the start and the width are
# written with: floating point makes 0.1 × 3 0.30000000000000004, which would put a
# value of 0.3 below its own boundary and count it a class low.
def decimals(x):
    """How many decimals x is written with, at 15 significant digits: 0.1 has 1, 20 has 0."""
    return max(0, -Decimal(f"{x:.15g}").as_tuple().exponent)


def boundary(start, width, i):
    return round(float(start + i * width), max(decimals(start), decimals(width)))


# The multiple of the width at or below the smallest value. The division can land a
# hair off a whole number (0.3 / 0.1 is 2.9999999999999996), so the boundaries decide.
def multiple_below(low, width):
    k = np.floor(low / width)
    if boundary(0.0, width, k + 1) <= low:
        k += 1
    elif boundary(0.0, width, k) > low:
        k -= 1
    return boundary(0.0, width, k)


# About 1 + log2(n) classes (Sturges' rule), with the width moved to the nearest of 1,
# 2.5, 5 or 10 times a power of ten -- nearest on a logarithmic scale, not rounded up.
# The first class starts at the multiple of the width at or below the smallest value.
# The range is worked out to the decimals the values are written with: floating point
# makes 1.1 - 0.6 0.5000000000000001, not 0.5.
def data_range(x):
    return round(float(x.max() - x.min()), max(decimals(x.min()), decimals(x.max())))


def first_class(x):
    target = max(1, int(np.ceil(1 + np.log2(x.size))))
    span = data_range(x) or 1.0
    magnitude = 10.0 ** np.floor(np.log10(span / target))
    steps = [round(m * magnitude, decimals(m * magnitude)) for m in (1, 2.5, 5, 10)]
    width = min(steps, key=lambda w: abs(np.log(w * target / span)))
    return multiple_below(x.min(), width), width


# Classes are [lower, upper): a value on a boundary belongs to the class above. Enough
# classes are added to hold the largest value, so the top boundary lies above it.
frequencies = {}
for level, x in series.items():
    start, width = first_class(x)
    count = 1
    while boundary(start, width, count) <= x.max():
        count += 1
    edges = np.array([boundary(start, width, i) for i in range(count + 1)])
    counts = np.histogram(x, bins=edges)[0]
    frequencies[level] = pd.DataFrame({"lower": edges[:-1], "frequency": counts,
                                       "relative": counts / x.size,
                                       "cumulative": np.cumsum(counts) / x.size})
    print(f"\n{level}: n = {x.size}, class width {width:g}")
    print(frequencies[level].to_string(index=False))
```

## Stem-and-leaf displays

```python
def stem_width(x):
    return float("5")


# The stem unit u is the power of ten at or above the width. A value is placed by its
# size |x|: the stem is |x| cut down to a multiple of u, and the leaf is the next digit,
# cut off rather than rounded, as in Tukey's displays: 3.97 is stem 3, leaf 9. A
# negative value takes the stem towards zero with a minus sign: -4.9 is stem -4, leaf 9,
# and a value between -1 and 0 sits on the stem -0. Each row holds the sizes in
# [lower, lower + width), its bounds on the decimal grid as the classes above are.
def power_of_ten(exponent):
    return round(10.0 ** exponent, max(0, -exponent))


def stem_unit(width):
    exponent = int(np.ceil(np.log10(width)))
    while power_of_ten(exponent - 1) >= width:
        exponent -= 1
    while power_of_ten(exponent) < width:
        exponent += 1
    return power_of_ten(exponent)


def stem_and_leaf(x, width):
    unit = stem_unit(width)
    leaf_unit = round(unit / 10, decimals(unit) + 1)
    rows_per_stem = round(unit / width)
    # Each value's row, counted from 0 on its size, and its leaf. x is sorted.
    placed = [(bool(v < 0), round(multiple_below(abs(v), width) / width),
               round(multiple_below(abs(v), leaf_unit) / leaf_unit) % 10) for v in x]
    (low_negative, low_row, _), (high_negative, high_row, _) = placed[0], placed[-1]
    # From the lowest value's row to the highest's: the negative rows from the largest
    # size down to the stem -0, then the stem 0 and up.
    order = []
    if low_negative:
        order += [(True, r) for r in range(low_row, (high_row if high_negative else 0) - 1, -1)]
    if not high_negative:
        order += [(False, r) for r in range(0 if low_negative else low_row, high_row + 1)]
    leaves = {key: "" for key in order}
    for negative, row, leaf in placed:
        leaves[(negative, row)] += str(leaf)
    rows = []
    for (negative, row), text in leaves.items():
        stem = round(row // rows_per_stem * unit, decimals(unit))
        rows.append([len(text), -stem if negative else stem, text])
    return pd.DataFrame(rows, columns=["frequency", "stem", "leaves"])


stems = {level: stem_and_leaf(x, stem_width(x)) for level, x in series.items()}
for level, table in stems.items():
    print(f"\n{level}")
    print(table.to_string(index=False))
```

## Box plot data

```python
# Tukey's rule: a whisker reaches the most extreme value lying within 1.5 × IQR of the
# box, and anything further out is listed as an outlier -- upper ones above the
# whisker (largest first), lower ones below it (smallest first). The whisker ends at
# a data value, not at the fence itself, and "outlier" is a drawing rule, not a test.
def box(x):
    q1, median, q3 = np.percentile(x, [25, 50, 75])
    low, high = q1 - 1.5 * (q3 - q1), q3 + 1.5 * (q3 - q1)
    inside = x[(x >= low) & (x <= high)]
    return {"upper whisker": inside.max(), "Q3": q3, "median": median, "Q1": q1,
            "lower whisker": inside.min(),
            "upper outliers": sorted(x[x > high], reverse=True),
            "lower outliers": sorted(x[x < low])}


boxes = {level: box(x) for level, x in series.items()}
most_upper = max(len(b["upper outliers"]) for b in boxes.values())
most_lower = max(len(b["lower outliers"]) for b in boxes.values())


def column(b):
    upper = b["upper outliers"] + [np.nan] * (most_upper - len(b["upper outliers"]))
    lower = b["lower outliers"] + [np.nan] * (most_lower - len(b["lower outliers"]))
    return upper + [b["upper whisker"], b["Q3"], b["median"], b["Q1"], b["lower whisker"]] + lower


labels = (["upper outlier"] * most_upper + ["upper whisker", "Q3", "median", "Q1", "lower whisker"]
          + ["lower outlier"] * most_lower)
boxdata = pd.DataFrame({level: column(b) for level, b in boxes.items()}, index=labels)
print()
print(boxdata)
```

## The numbers behind the mean plots

```python
means = summary.loc["mean"].to_numpy(float)
sds = summary.loc["sd"].to_numpy(float)
ses = summary.loc["se"].to_numpy(float)
band = pd.DataFrame([means + sds, means + ses, means, means - ses, means - sds],
                    index=["mean + SD", "mean + SE", "mean", "mean - SE", "mean - SD"],
                    columns=levels)
error_se = pd.DataFrame([means, ses], index=["mean", "SE"], columns=levels)
error_sd = pd.DataFrame([means, sds], index=["mean", "SD"], columns=levels)
print()
print(band)
print()
print(error_se)
print()
print(error_sd)
```

## Graphs

```python
# The same values drawn four ways, each keeping something different: the box plot
# keeps the centre, the spread and the outliers; mean ± SE keeps only how precisely
# each mean is known, and hides every single value; the dot plot keeps every value;
# the kernel density (stats.gaussian_kde, Scott's bandwidth SD × n^(-1/5), drawn three
# bandwidths beyond the data) keeps the shape. Pair a summary view with one that
# shows the raw values, so nothing important is averaged away.
Path("figures").mkdir(exist_ok=True)
fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes[0, 0].bxp([{"label": level, "med": b["median"], "q1": b["Q1"], "q3": b["Q3"],
                 "whislo": b["lower whisker"], "whishi": b["upper whisker"],
                 "fliers": b["lower outliers"] + b["upper outliers"]}
                for level, b in boxes.items()])
axes[0, 1].errorbar(range(len(levels)), means, yerr=ses, fmt="o", capsize=6)
axes[0, 1].set_xticks(range(len(levels)), levels)
for row, (level, x) in enumerate(series.items()):
    seen = {}
    heights = []
    for v in x:
        heights.append(row + 0.06 * seen.get(v, 0))
        seen[v] = seen.get(v, 0) + 1
    axes[1, 0].scatter(x, heights, s=10)
    kde = stats.gaussian_kde(x)
    bandwidth = kde.factor * np.std(x, ddof=1)
    grid = np.linspace(x.min() - 3 * bandwidth, x.max() + 3 * bandwidth, 100)
    axes[1, 1].plot(grid, kde(grid), label=level)
axes[1, 0].set_yticks(range(len(levels)), levels)
axes[1, 1].legend()
fig.tight_layout()
fig.savefig("figures/statistical_graphs.png", dpi=150)
plt.close(fig)
```

## What the result says

The table has one row per order and an App column. One run gives the descriptive statistics, frequency tables, stem-and-leaf displays, box plot, mean plots, density curves and dot plot for each app.

Every view tells the same story. ZipEats is fastest, with a mean (average) of 24.46 minutes, and UrbanBite is slowest, with 39.76. The stem-and-leaf display and the box plot both show ZipEats' lone order of 58 minutes.

The coefficient of variation is the standard deviation divided by the mean. It is 0.265 for ZipEats and 0.128 for FeastNow, so ZipEats looks twice as erratic. But the interquartile range, the spread of the middle half of the orders, is 5 minutes for both. One slow order causes the difference.
