These cells save `grubbs.py`: a step that no library does, written out. The analysis imports it.

```python
%%writefile grubbs.py
"""The Smirnov-Grubbs test for one outlier. SciPy and statsmodels have no such test.

G is how many standard deviations the candidate sits from the mean. Its P value comes
from Student's t, Bonferroni-corrected over the values that could have been the
extreme one: 2n under a two-sided test (either end could have been picked), n under a
one-sided test (the end was named in advance, so the candidate is always that end's
value).
"""

import numpy as np
from scipy import stats


def grubbs(x, side="two-sided"):
    """(position of the candidate in x, G, P). side: "two-sided", "largest" or "smallest"."""
    n = x.size
    mean, sd = np.mean(x), np.std(x, ddof=1)
    if sd == 0:
        return -1, 0.0, 1.0
    deviations = x - mean
    if side == "two-sided":
        index = int(np.argmax(np.abs(deviations)))
    elif side == "largest":
        index = int(np.argmax(deviations))
    else:
        index = int(np.argmin(deviations))
    g = abs(deviations[index]) / sd
    denominator = (n - 1) ** 2 - n * g**2
    if denominator <= 0:
        return index, g, 0.0
    t = np.sqrt(n * (n - 2) * g**2 / denominator)
    eligible = 2 * n if side == "two-sided" else n
    return index, g, min(1.0, eligible * stats.t.sf(t, n - 2))
```

## Libraries

```python
import numpy as np
import pandas as pd
from grubbs import grubbs

pd.set_option("display.width", 140)
```

## The data

```python
# One column per variable. Blank and text cells are skipped but keep their row
# number, so a result can be traced back to its row.
# Every test below assumes the values other than the candidate come from one normal
# distribution. A small P says the candidate sits further out than such a sample
# would plausibly put its most extreme value; it cannot say why (a typo, a fault, a
# genuine rare case), so check the record before deleting anything, and report the
# results with and without it. Testing one value at a time can miss two outliers
# lying close together, which inflate the SD between them (masking).
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
columns = ["Price ($)"]
alpha = 5 / 100
# Two-sided: the candidate is whichever value lies furthest from the mean.
side = "two-sided"
```

## Repeated Smirnov-Grubbs test

```python
# Test the candidate (grubbs.py); if P < alpha, exclude it and test again on what is
# left, until a candidate is retained or fewer than 3 values remain. Rows are tracked
# by position, so of two equal values only the excluded one is marked.
# Smirnov-Grubbs allows for having picked the most extreme of n values (a Bonferroni
# correction), so it is the standard choice for "is the most extreme value an outlier?".
results = {}
for name in columns:
    raw = pd.to_numeric(data[name], errors="coerce").to_numpy(float)
    alive = np.flatnonzero(~np.isnan(raw))
    total = alive.size
    removed, steps = [], []
    while alive.size >= 3:
        x = raw[alive]
        index, statistic, p = grubbs(x, side)
        # When every value is the same, grubbs() returns -1: nothing can be an outlier.
        if index < 0:
            break
        is_max = x[index] >= x.mean() if side == "two-sided" else side == "largest"
        rejected = p < alpha
        steps.append([len(steps) + 1, x.size, x.mean(), x.var(ddof=1), x.std(ddof=1),
                      np.nan if is_max else x.min(), x.max() if is_max else np.nan,
                      alive[index] + 1, p, "rejected" if rejected else "retained",
                      "largest is an outlier" if is_max else "smallest is an outlier"])
        if not rejected:
            break
        removed.append(alive[index])
        alive = np.delete(alive, index)

    outlier = np.isin(np.arange(raw.size), removed)
    results[name] = {
        "data": pd.DataFrame({"No.": np.arange(1, raw.size + 1), "original": raw,
                              "after exclusion": np.where(outlier, np.nan, raw),
                              "outlier": np.where(outlier, raw, np.nan)}),
        "summary": pd.Series({"original": total, "after exclusion": total - len(removed),
                              "outliers": len(removed)}),
        "trace": pd.DataFrame(steps, columns=["step", "n", "mean", "variance", "sd", "min",
                                              "max", "No.", "P", "verdict", "alternative"]),
    }
    print(f"\n=== {name} ===")
    print(results[name]["data"].to_string(index=False))
    print()
    print(results[name]["summary"].to_string())
    print()
    print(results[name]["trace"].to_string(index=False))
```

## What the result says

The Smirnov-Grubbs test asks whether the most extreme value is too far from the mean (the average) to come from the same bell curve as the rest. Two listings have no price, so 45 prices are tested.

In step 1 the largest price, \$1,999, is rejected. Its P value, the chance of a value this extreme if nothing were wrong, is practically zero. In step 2 the test looks at the 44 prices that remain. The most extreme is now the smallest, \$130, and it is retained with P = 1. No other price is suspicious.

Without the \$1,999 listing the mean price falls from about \$204 to about \$163. The test cannot say what that listing is, so deleting it is a business decision.
