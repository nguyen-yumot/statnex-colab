## Libraries

```python
import numpy as np
import pandas as pd

pd.set_option("display.width", 160)
```

## The data

```python
# One column per variable, one row per case. A case with a blank or non-numeric
# value in any of the variables is dropped from all of them.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
columns = ["Temperature (C)", "Visitors (thousands)", "Ice cream sales ($000)", "Sunscreen sales ($000)", "Hot coffee sales ($000)", "Rainy days"]
numeric = data[columns].apply(pd.to_numeric, errors="coerce").dropna()
```

## Variance-covariance matrix

```python
# The diagonal holds each variable's variance (its spread, in squared units); each
# off-diagonal cell is a covariance: positive when the two tend to be above their
# means together, negative when one is high as the other is low. Its size depends on
# both variables' units, so compare signs across pairs, and use correlations to
# compare strength.
# Each sum of cross-products is divided by n, treating the cases as the whole
# population. (NumPy's np.cov and pandas' cov() divide by n - 1 unless told otherwise.)
ddof = 0
covariance = pd.DataFrame(np.cov(numeric.to_numpy(float), rowvar=False, ddof=ddof),
                          index=columns, columns=columns)
sizes = pd.DataFrame(len(numeric), index=columns, columns=columns)
print(covariance)
print(f"\nn = {len(numeric)}")
```

## What the result says

The diagonal of the matrix holds each measure's variance, which is its spread in squared units. Temperature has the largest, 66.4, and rainy days the smallest, 5.06. The other cells hold covariances. A covariance shows whether two measures move in the same direction (plus sign) or in opposite directions (minus sign).

Temperature and ice cream sales have a covariance of 46.4, and temperature and sunscreen 31.9: warm months sell more of both. Temperature and hot coffee have -41.4: warm months sell less coffee.

Read the signs, not the sizes. A covariance grows with the units and the swings of each measure, so 46.4 against 31.9 does not mean that ice cream has the stronger link. To compare strength, use correlations.
