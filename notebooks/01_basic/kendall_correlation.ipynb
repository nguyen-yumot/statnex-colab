## Libraries

```python
from itertools import combinations

import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.width", 200)
```

## The data

```python
# One column per variable, one row per case. A case with a blank or non-numeric
# value in any of the variables is dropped from all of them.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
columns = ["Price tier", "Sweetness", "Saltiness", "Crunchiness", "Rich flavour", "Healthy image", "Portion size", "Packaging appeal", "Brand awareness", "Novelty", "Shareability", "Kid appeal", "Value for money", "Repurchase intent"]
numeric = data[columns].apply(pd.to_numeric, errors="coerce").dropna()
```

## Kendall's rank correlation (tau-b)

```python
# Over every pair of cases: do the two variables agree about which case is larger
# (concordant) or disagree (discordant)? τ is concordant minus discordant pairs,
# scaled so that ties do not pull it towards 0 (tau-b). It uses only the order of the
# values, so it suits ranks and banded scores, and it is unchanged by any
# transformation that keeps the order. Without ties, τ = 0.5 means 75% of pairs agree
# and 25% disagree. SciPy's P value (for "τ = 0") is exact when neither variable has
# ties and n <= 33, and otherwise uses the normal approximation with a correction for
# ties and no continuity correction.
tau = pd.DataFrame(np.eye(len(columns)), index=columns, columns=columns)
p_values = pd.DataFrame(np.nan, index=columns, columns=columns)
for a, b in combinations(columns, 2):
    result = stats.kendalltau(numeric[a], numeric[b])
    tau.loc[a, b] = tau.loc[b, a] = result.statistic
    p_values.loc[a, b] = p_values.loc[b, a] = result.pvalue
sizes = pd.DataFrame(len(numeric), index=columns, columns=columns)
print(tau.round(4))
```

## Test of the rank correlation coefficients

```python
# The app prints the P value above the diagonal and * (P < 0.05) / ** (P < 0.01)
# below it. The tests assume independent cases. A matrix of k variables runs
# k(k - 1)/2 tests, and about 1 in 20 of them shows P < 0.05 by chance alone, so a
# lone starred pair is weak evidence; read τ for the strength.
print()
print(p_values.round(4))
print(f"\nn = {len(numeric)}")
```

## What the result says

Kendall's tau-b is a rank correlation. It runs from -1 to 1 and shows whether two rankings rise together (plus) or move in opposite directions (minus). Price tier goes with packaging appeal (tau = 0.75). It goes against value for money (-0.68) and portion size (-0.58). For all three the P value, the chance of a link this strong if there were none, is practically zero.

For price tier and repurchase intent, tau is -0.15 with P = 0.163. A link this weak can easily come from chance in 56 snacks.

So premium snacks come with better packaging, smaller portions and worse value for money. The panel does not show that they earn more repeat purchase, and it does not prove that they earn the same.
