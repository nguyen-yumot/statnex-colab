# Product displays in five stores: a blocked plan

**Randomized Block Design** · a worked example from the StatNex teaching guide

**The story.** A retailer wants to test four ways of displaying a new product (window display, entrance table, end-cap shelf, checkout stand) in five stores, one display a week for four weeks. Stores differ a lot in size and customers, so each store is treated as a block.

**The question.** In what order should each store try the four displays?

## Libraries

```python
import numpy as np
import pandas as pd

pd.set_option("display.width", 120)
```

## The treatments

```python
# The conditions to compare, one per row of a single column.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
treatments = [str(value).strip() for value in data["Display layout"].dropna()
              if str(value).strip()]
k = len(treatments)
blocks = 5
print(f"{k} treatments x {blocks} blocks = {k * blocks} runs")
```

## Random numbers

```python
# NumPy's generator with the same seed as the app, so the plans below are the app's.
# A seed of 0 means no seed: a different plan every time the script runs.
seed = 20260809
rng = np.random.default_rng(seed if seed else None)
```

## Randomized block design

```python
# Each block (a column) gets every treatment exactly once, in its own random order; the
# rows are the order of the runs. Whatever differs between blocks then affects every
# treatment equally and drops out of the comparison.
blocked = pd.DataFrame({f"round {b + 1}": rng.permutation(treatments) for b in range(blocks)})
print()
print(blocked.to_string(index=False))
```

## Random sampling, for comparison

```python
# The same number of runs with each one drawn freely from all treatments (with
# replacement): some come up twice in a round and some not at all, and the totals need
# not even be equal. A completely randomised design would instead shuffle the same
# number of copies of each treatment over all the runs, ignoring the blocks.
free = pd.DataFrame({f"round {b + 1}": [treatments[i] for i in rng.integers(0, k, k)]
                     for b in range(blocks)})
print()
print(free.to_string(index=False))
print("\nruns per treatment under complete randomisation:")
print(pd.Series(free.to_numpy().ravel()).value_counts().reindex(treatments, fill_value=0))
```

## What the result says

This method makes a plan, not a test. In a randomized block design, each block (here a store) receives every treatment (here a display) once, in a random order. Each Round column is one store and each row is one week.

In the randomized block design table, every column holds all four displays. So every display is tried in every store.

The random sampling table shows what happens without blocks. Round 5 has the end-cap shelf in all four weeks, and Round 3 never has the checkout stand. With such a plan, a display could look good only because it was in a busy store. Use the blocked plan.
