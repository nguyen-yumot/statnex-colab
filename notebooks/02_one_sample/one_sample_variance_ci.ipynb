## Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats

pd.set_option("display.width", 120)
pd.set_option("display.float_format", "{:.6g}".format)
```

## The data

```python
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
variables = ["Delivery time (min)", "Distance (km)", "Order total ($)", "Tip ($)"]
confidence = 95 / 100
```

## Confidence interval for the variance, one column at a time

```python
# (n - 1) s² / σ² follows a chi-square distribution with n - 1 degrees of freedom, so
# the interval is built for the variance and printed on the standard-deviation scale
# by taking square roots. It is not symmetric: it reaches further above s than below.
# The interval is exact only for normal data, and unlike an interval for a mean it does
# not become safe as n grows: a long tail or one extreme value makes it too narrow.
# Plot the data first; if the tails are heavy, a robust interval (Bonett's) or a
# bootstrap is the safer report.
results = {}
for name in variables:
    # Text and blank cells count as missing and are skipped.
    x = pd.to_numeric(data[name], errors="coerce").dropna().to_numpy(float)
    n, sd = len(x), x.std(ddof=1)
    df = n - 1
    variance = sd**2
    # The upper chi-square quantile gives the lower limit, and the lower one the upper.
    lower = df * variance / stats.chi2.ppf(0.5 + confidence / 2, df)
    upper = df * variance / stats.chi2.ppf(0.5 - confidence / 2, df)
    results[name] = {"n": n, "variance": variance, "lower (SD)": np.sqrt(lower),
                     "SD": sd, "upper (SD)": np.sqrt(upper)}

table = pd.DataFrame(results)
print(table)
print(f"\nConfidence level: {confidence:.1%}")
```

## What the result says

The standard deviation of the 50 delivery times is 6.49 minutes: a typical delivery lands about that far from the average. The variance, 42.1, is the standard deviation squared.

A sample gives only an estimate of the spread, so the result adds a 95% confidence interval, a range of values the data agree with for the true standard deviation of all orders. Built from the chi-square distribution, it runs from 5.42 to 8.09 minutes. For the order total, the standard deviation is \$3.77, with an interval from \$3.15 to \$4.70.

Even that range is too optimistic for delivery times. This interval assumes bell-shaped data, and one extremely late order shows that delivery times are not bell-shaped. The real uncertainty about their spread is larger.
