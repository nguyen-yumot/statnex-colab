```python
%%writefile collinear_columns.py
"""Which explanatory variables are exact linear combinations of the ones before them.

Such a variable adds no information and makes the model impossible to estimate. Each
column is centred and scaled to length 1, then regressed on the columns already kept;
if nothing is left over (squared residual length at most 1e-10) the column is
redundant. A constant column is redundant too. The *later* member of a redundant pair
is the one reported, so a variable listed first survives.
"""

import numpy as np


def collinear_columns(values, names, tolerance=1e-10):
    centred = values - values.mean(axis=0)
    scale = np.linalg.norm(centred, axis=0)
    scale[scale == 0] = 1.0
    centred = centred / scale
    dropped, kept = [], []
    for j, name in enumerate(names):
        column = centred[:, j]
        if kept:
            basis = centred[:, kept]
            column = column - basis @ np.linalg.lstsq(basis, column, rcond=None)[0]
        if column @ column <= tolerance or values[:, j].max() == values[:, j].min():
            dropped.append(name)
        else:
            kept.append(j)
    return dropped
```

```python
%%writefile mixed_model.py
"""Maximum likelihood for a one-level linear mixed model, to the last digit.

Each group's observations have covariance V = Z T Z' + s2 I: T is the covariance of
the group's random effects, s2 the residual variance. For given (T, s2) the fixed
effects come from generalised least squares, so the likelihood is a function of the
covariance parameters alone.

statsmodels' MixedLM gives the starting point. Its optimiser cannot stop on the edge
of the parameter space, where T is singular (some combination of the random effects
has variance 0), and there it can stop far short of the maximum. `profile_fit` then
searches as lme4 does (Bates et al., 2015): over the Cholesky factor of T / s2, with
its diagonal kept at 0 or above and s2 worked out exactly, so the edge is reachable.
`fisher_scoring` finishes the job inside the space: it steps by the expected
information until the score is zero to working precision. The same information gives
the standard errors of T and s2, which MixedLM does not print in this form.
"""

import numpy as np
from scipy import optimize, stats


def gls(y, X, Z, groups, T, s2):
    """Fixed effects, -2 log likelihood, and X'V⁻¹X for one (T, s2)."""
    xtvx = np.zeros((X.shape[1], X.shape[1]))
    xtvy = np.zeros(X.shape[1])
    inverses, log_det = [], 0.0
    for index in groups:
        V = Z[index] @ T @ Z[index].T + s2 * np.eye(len(index))
        inverse = np.linalg.inv(V)
        inverses.append(inverse)
        log_det += np.linalg.slogdet(V)[1]
        xtvx += X[index].T @ inverse @ X[index]
        xtvy += X[index].T @ inverse @ y[index]
    beta = np.linalg.solve(xtvx, xtvy)
    quadratic = sum((y[i] - X[i] @ beta) @ inv @ (y[i] - X[i] @ beta)
                    for i, inv in zip(groups, inverses))
    return beta, log_det + quadratic + len(y) * np.log(2 * np.pi), xtvx, inverses


def profile_fit(y, X, Z, groups, T, s2):
    """Maximum likelihood over T and s2, started at (T, s2), as lme4 searches.

    T / s2 = L L' with L lower triangular and a diagonal of 0 or more; for a given L the
    fixed effects and s2 have closed forms, so the search is over L alone (L-BFGS-B).
    Returns T, s2 and whether the fit is singular: a diagonal element of L below 1e-4,
    lme4's rule in isSingular().
    """
    n, q = len(y), Z.shape[1]
    lower = np.tril_indices(q)

    def unpack(theta):
        factor = np.zeros((q, q))
        factor[lower] = theta
        return factor @ factor.T

    def deviance(theta):
        relative = unpack(theta)
        beta, _, _, inverses = gls(y, X, Z, groups, relative, 1.0)
        log_det = sum(np.linalg.slogdet(Z[i] @ relative @ Z[i].T + np.eye(len(i)))[1]
                      for i in groups)
        quadratic = sum((y[i] - X[i] @ beta) @ inverse @ (y[i] - X[i] @ beta)
                        for i, inverse in zip(groups, inverses))
        return n * np.log(quadratic / n) + log_det + n * (1 + np.log(2 * np.pi)), quadratic / n

    start = np.linalg.cholesky(T / s2 + 1e-10 * np.eye(q))[lower]
    bounds = [(0, None) if a == b else (None, None) for a, b in zip(*lower)]
    result = optimize.minimize(lambda theta: deviance(theta)[0], start, method="L-BFGS-B",
                               bounds=bounds,
                               options={"ftol": 1e-15, "gtol": 1e-12, "maxiter": 10000,
                                        "maxfun": 100000})
    s2 = deviance(result.x)[1]
    on_diagonal = lower[0] == lower[1]
    factor = np.zeros((q, q))
    factor[lower] = result.x * np.sqrt(s2)
    singular = bool(result.x[on_diagonal].min() < 1e-4)
    if singular:
        # On the edge the likelihood is nearly flat, and the search stops wherever its
        # tolerance says, on the edge or just short of it. Holding the near-zero diagonal
        # elements at 0, Fisher scoring on the edge pins the maximum down; it is kept if
        # it does at least as well as the search.
        edge = factor.copy()
        edge[np.diag_indices(q)] = np.where(result.x[on_diagonal] < 1e-4, 0.0, np.diag(factor))
        edge, edge_s2 = edge_scoring(y, X, Z, groups, edge, s2)
        if gls(y, X, Z, groups, edge @ edge.T, edge_s2)[1] <= gls(y, X, Z, groups, factor @ factor.T, s2)[1]:
            factor, s2 = edge, edge_s2
    return factor @ factor.T, s2, singular


def edge_scoring(y, X, Z, groups, factor, s2, rounds=200):
    """Fisher scoring on the edge, as the app does: in the Cholesky factor C of T,
    with the diagonal elements that reached 0 held there. T's score and information
    are carried over by dT_ij/dC_ab = δ_ia C_jb + δ_ja C_ib."""
    q = Z.shape[1]
    cells = list(zip(*np.tril_indices(q)))
    free = [(a, b) for a, b in cells if not (a == b and factor[a, a] == 0.0)]
    for _ in range(rounds):
        jacobian = np.zeros((len(cells) + 1, len(free) + 1))
        for column, (a, b) in enumerate(free):
            for row, (i, j) in enumerate(cells):
                jacobian[row, column] = ((factor[j, b] if i == a else 0.0)
                                         + (factor[i, b] if j == a else 0.0))
        jacobian[-1, -1] = 1.0
        score, information = score_information(y, X, Z, groups, factor @ factor.T, s2)
        score, information = jacobian.T @ score, jacobian.T @ information @ jacobian
        step = np.linalg.lstsq(information, score, rcond=None)[0]
        if not np.all(np.isfinite(step)) or abs(score @ step) < 1e-24:
            break
        current = gls(y, X, Z, groups, factor @ factor.T, s2)[1]
        for shrink in (1.0, 0.5, 0.25, 0.125):
            candidate = factor.copy()
            for k, (a, b) in enumerate(free):
                candidate[a, b] += shrink * step[k]
            new_s2 = s2 + shrink * step[-1]
            if new_s2 > 0 and gls(y, X, Z, groups, candidate @ candidate.T, new_s2)[1] <= current + 1e-10:
                factor, s2 = candidate, new_s2
                break
        else:
            break
    return factor, s2


def _derivatives(Z_g, q):
    """dV/d(parameter) for each lower-triangle element of T, then for s2."""
    out = []
    for a, b in zip(*np.tril_indices(q)):
        selector = np.zeros((q, q))
        selector[a, b] = selector[b, a] = 1.0
        out.append(Z_g @ selector @ Z_g.T)
    out.append(np.eye(len(Z_g)))
    return out


def score_information(y, X, Z, groups, T, s2):
    """The score and the expected (Fisher) information of (T's lower triangle, s2)."""
    q = Z.shape[1]
    beta, _, _, inverses = gls(y, X, Z, groups, T, s2)
    size = q * (q + 1) // 2 + 1
    score, information = np.zeros(size), np.zeros((size, size))
    for index, inverse in zip(groups, inverses):
        residual = inverse @ (y[index] - X[index] @ beta)
        products = [inverse @ d for d in _derivatives(Z[index], q)]
        for i, (d_i, p_i) in enumerate(zip(_derivatives(Z[index], q), products)):
            score[i] += -0.5 * (np.trace(p_i) - residual @ d_i @ residual)
            for j, p_j in enumerate(products):
                information[i, j] += 0.5 * np.trace(p_i @ p_j)
    return score, information


def fisher_scoring(y, X, Z, groups, T, s2, rounds=60):
    """Step by information⁻¹ · score until the score vanishes; halve a step that would
    leave T not positive definite or raise -2 log L."""
    lower = np.tril_indices(Z.shape[1])
    for _ in range(rounds):
        score, information = score_information(y, X, Z, groups, T, s2)
        if np.abs(score).max() < 1e-11:
            break
        step = np.linalg.solve(information, score)
        current = gls(y, X, Z, groups, T, s2)[1]
        for shrink in (1.0, 0.5, 0.25, 0.125):
            candidate = T.copy()
            candidate[lower] += shrink * step[:-1]
            candidate = np.tril(candidate) + np.tril(candidate, -1).T
            new_s2 = s2 + shrink * step[-1]
            if (new_s2 > 0 and np.linalg.eigvalsh(candidate).min() > 0
                    and gls(y, X, Z, groups, candidate, new_s2)[1] <= current + 1e-9):
                T, s2 = candidate, new_s2
                break
        else:
            break
    return T, s2


def covariance_errors(y, X, Z, groups, T, s2):
    """Standard errors of T (as a symmetric matrix) and of s2, from the information."""
    q = Z.shape[1]
    variance = np.diag(np.linalg.inv(score_information(y, X, Z, groups, T, s2)[1]))
    errors = np.sqrt(np.clip(variance, 0, None))
    matrix = np.zeros((q, q))
    matrix[np.tril_indices(q)] = errors[:-1]
    return matrix + np.tril(matrix, -1).T, errors[-1]


def homogeneity(y, Z, groups, gamma, s2):
    """Raudenbush and Bryk's chi-square that each random coefficient really varies.

    Every group with more rows than random effects is fitted on its own by least
    squares; the spread of those estimates around the fixed effect is compared with
    the sampling error they would have if every group shared one value (using the
    pooled s2). Degrees of freedom: usable groups - 1.
    """
    estimates, precisions = [], []
    for index in groups:
        if len(index) <= Z.shape[1]:
            continue
        inverse = np.linalg.inv(Z[index].T @ Z[index])
        estimates.append(inverse @ Z[index].T @ y[index])
        precisions.append(1 / (s2 * np.diag(inverse)))
    estimates, precisions = np.array(estimates), np.array(precisions)
    chi2 = (precisions * (estimates - gamma) ** 2).sum(axis=0)
    df = len(estimates) - 1
    return chi2, df, stats.chi2.sf(chi2, df)
```

```python
# One row per observation; the stratum column says which group (level-2 unit) it is in.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
response = "Job satisfaction"
predictors = ["Work-life balance", "Pay level"]
stratum = "Company"
# Explanatory variables whose slope also varies from group to group.
random_names = ["Work-life balance", "Pay level"]
frame = data[[response, *predictors]].apply(pd.to_numeric, errors="coerce")
max_iter = 50
confidence = 95 / 100


# A stratum as text; a number read as 2.0 prints as "2".
def label(value):
    return str(int(value)) if isinstance(value, float) and value.is_integer() else str(value)
```

## Cases

```python
r, x_ok, s = frame[response].notna(), frame[predictors].notna().all(axis=1), data[stratum].notna()
summary = pd.DataFrame({"n": [
    (r & x_ok & s).sum(), (~r & x_ok & s).sum(), (r & ~x_ok & s).sum(), (r & x_ok & ~s).sum(),
    (r & ~x_ok & ~s).sum(), (~r & x_ok & ~s).sum(), (~r & ~x_ok & s).sum(),
    (~r & ~x_ok & ~s).sum(), len(frame),
]}, index=["valid", "response missing", "explanatory missing", "stratum missing",
           "only the response", "only the explanatory", "only the stratum", "all missing", "total"])
summary["%"] = summary["n"] / len(frame)
print(summary)
valid = frame[r & x_ok & s]
y = valid[response].to_numpy(float)
x = valid[predictors].to_numpy(float)
strata = data.loc[r & x_ok & s, stratum].map(label).to_numpy()
# Strata in the order they first appear.
levels = list(dict.fromkeys(strata))
codes = np.array([levels.index(v) for v in strata])
groups = [np.flatnonzero(codes == k) for k in range(len(levels))]
n = len(y)

response_summary = pd.DataFrame({"n": [*np.bincount(codes), n]}, index=[*levels, "total"])
response_summary["%"] = response_summary["n"] / n
descriptive = pd.DataFrame({"n": n, "mean": valid.mean(), "variance": valid.var(ddof=1),
                            "SD": valid.std(ddof=1), "min": valid.min(), "max": valid.max()}
                           ).loc[[*predictors, response]]
correlation_matrix = valid[[*predictors, response]].corr()
print()
print(response_summary)
print()
print(descriptive)
print()
print(correlation_matrix)
```

## Centring

```python
# Each explanatory variable minus its overall mean, so the constant is the response at
# the average of every variable.
x = x - x.mean(axis=0)
```

## Variables that are linear combinations of others

```python
dropped = collinear_columns(x, predictors)
print("\nLinearly dependent:", dropped or "none")
# They are left out of the model.
keep = [j for j, name in enumerate(predictors) if name not in dropped]
x, predictors = x[:, keep], [predictors[j] for j in keep]
```

## The model

```python
# Fixed effects: every explanatory variable and a constant (last, as the app prints it).
# Random effects: the chosen slopes, then the constant if the intercept varies.
X = np.column_stack([x, np.ones(n)])
names = [*predictors, "Constant"]
random_columns = [predictors.index(name) for name in random_names]
random_columns.append(len(predictors))
Z = X[:, random_columns]
random_labels = [names[j] for j in random_columns]


# Maximum likelihood (not REML, so that models with different fixed effects can be
# compared by their -2 log likelihood). statsmodels' MixedLM gives the start; the search
# then continues as lme4's does, which can reach the edge where T is singular
# (profile_fit), and Fisher scoring finishes it until the score is zero.
def maximum_likelihood(X, Z):
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        start = MixedLM(y, X, groups=codes, exog_re=Z).fit(reml=False, method="lbfgs",
                                                             maxiter=max(max_iter, 200))
    T, s2, singular = profile_fit(y, X, Z, groups, np.asarray(start.cov_re), float(start.scale))
    return (*fisher_scoring(y, X, Z, groups, T, s2), singular)


T, s2, singular = maximum_likelihood(X, Z)
beta, deviance, xtvx, _ = gls(y, X, Z, groups, T, s2)
# A singular T (a "boundary fit") means one combination of the random effects has
# variance 0: the data cannot support this many random effects. The estimates are still
# the maximum, but think about dropping a random slope whose variance is near 0.
if singular:
    print("Boundary (singular) fit: the covariance matrix of the random effects is singular.")

# The null model drops every explanatory variable but keeps the random structure
# (the random intercept, or else the first random slope).
null_X = np.ones((n, 1))
null_Z = np.ones((n, 1))
null_deviance = gls(y, null_X, null_Z, groups, *maximum_likelihood(null_X, null_Z)[:2])[1]
q = Z.shape[1]
# Parameters counted for AIC and BIC: the fixed effects, T's distinct elements and s2;
# the null model has a constant, one random-effect variance and s2.
k_full = X.shape[1] + q * (q + 1) // 2 + 1
k_null = 3
chi2, df = null_deviance - deviance, k_full - k_null
fit = pd.DataFrame(
    [[null_deviance, null_deviance + 2 * k_null, null_deviance + k_null * np.log(n),
      np.nan, np.nan, np.nan],
     [deviance, deviance + 2 * k_full, deviance + k_full * np.log(n), chi2, df,
      stats.chi2.sf(chi2, df)]],
    index=["without the variables", "with the variables"],
    columns=["-2LL", "AIC", "BIC", "chi-square", "df", "P"])
print()
print(fit)
```

## Fixed effects

```python
# t tests on the between-groups degrees of freedom (groups - 1): the groups, not the
# observations, are the independent units.
df_fixed = len(levels) - 1
se = np.sqrt(np.diag(np.linalg.inv(xtvx)))
critical = stats.t.ppf(0.5 + confidence / 2, df_fixed)
fixed = pd.DataFrame({"b": beta, "SE": se, "lower": beta - critical * se,
                      "upper": beta + critical * se, "t": beta / se, "df": df_fixed,
                      "P": 2 * stats.t.sf(np.abs(beta / se), df_fixed)}, index=names)
print()
print(fixed)
```

## Random effects

```python
chi2_q, df_q, p_q = homogeneity(y, Z, groups, beta[random_columns], s2)
random = pd.DataFrame({"SD": np.sqrt(np.diag(T)), "variance": np.diag(T), "chi-square": chi2_q,
                       "df": df_q, "P": p_q}, index=random_labels)
random.loc["residual"] = [np.sqrt(s2), s2, np.nan, np.nan, np.nan]
covariance = pd.DataFrame(T, index=random_labels, columns=random_labels)
spread = np.sqrt(np.diag(T))
correlation = covariance / np.outer(spread, spread)
covariance_se, residual_se = covariance_errors(y, X, Z, groups, T, s2)
covariance_se = pd.DataFrame(covariance_se, index=random_labels, columns=random_labels)
print()
print(random)
print()
print(covariance)
print("residual variance:", s2)
print()
print(correlation)
print()
print(covariance_se)
print("standard error of the residual variance:", residual_se)
```

## Intraclass correlation

```python
# The constant's variance component over itself plus the residual variance: the share of
# the variance that lies between strata, from the fitted model as it stands. The constant
# is the last random effect. With random slopes, it is the value where they are 0.
icc = [T[-1, -1], s2, T[-1, -1] / (T[-1, -1] + s2)]
print()
print("intraclass correlation:", icc[2])
```

## What the result says

A linear mixed model is a regression for grouped data, here 200 employees in 40 companies. The fixed effects are the average slopes. The random effects let each company have its own level and its own slopes.

The fixed effects: one more step of work-life balance goes with 2.63 more satisfaction points, and one more step of pay with 2.30. Both P values are almost zero, so chance alone would hardly give such slopes, and both confidence intervals (the ranges of values the data fit) stay well above zero: 2.16 to 3.10 and 1.93 to 2.68.

The variance components describe the companies. The company level has a variance of 47.3, next to a residual variance of 47.2 inside companies. So the intraclass correlation, the share of the unexplained variation that lies between companies, is 0.50. So balance and pay both go with happier staff, and the company itself matters as much as anything.
