```python
%%writefile factor_analysis_ml.py
"""Maximum-likelihood factor extraction, by the profile likelihood over the uniquenesses.

statsmodels' Factor(method="ml") fits the same model but stops at a looser optimum, so
the recipe is written out here, the same way StatNex does it.

For any uniquenesses psi (each item's error variance) the best loadings follow in
closed form from the eigenvectors of psi^(-1/2) R psi^(-1/2), so only the p uniquenesses
are searched over. The criterion is

    F(psi) = sum over the discarded eigenvalues theta of (theta - log theta - 1),

which is zero when everything the factors leave out is pure noise (theta = 1). Its
gradient is diag(LL' + psi - R) / psi^2, so at the optimum the model reproduces the
diagonal exactly. L-BFGS-B gets close; a root solve on the gradient finishes the job.
"""

import warnings

import numpy as np
from scipy.optimize import minimize, root


def ml_loadings(correlation, factors, start, max_iterations):
    """(loadings, L-BFGS-B iterations, converged) from starting communalities ``start``."""
    p = correlation.shape[0]

    def spectrum(psi):
        scale = 1 / np.sqrt(psi)
        eigenvalues, vectors = np.linalg.eigh(correlation * np.outer(scale, scale))
        order = np.argsort(eigenvalues)[::-1]
        return eigenvalues[order], vectors[:, order]

    def loadings_at(psi):
        eigenvalues, vectors = spectrum(psi)
        kept = np.sqrt(np.clip(eigenvalues[:factors] - 1, 0.0, None))
        return np.sqrt(psi)[:, None] * (vectors[:, :factors] * kept)

    def objective(psi):
        rest = spectrum(psi)[0][factors:]
        return float(np.sum(rest - np.log(rest)) - (p - factors))

    def gradient(psi):
        loadings = loadings_at(psi)
        return np.diag(loadings @ loadings.T + np.diag(psi) - correlation) / psi**2

    floor = 1e-8
    guess = np.clip(1 - start, floor, 1.0)
    with warnings.catch_warnings():
        warnings.simplefilter("ignore")
        search = minimize(objective, guess, jac=gradient, method="L-BFGS-B",
                          bounds=[(floor, 1.0)] * p,
                          options={"ftol": 1e-18, "gtol": 1e-14,
                                   "maxiter": max(max_iterations, 200)})
        polished = root(gradient, np.clip(search.x, floor, 1.0), method="hybr", tol=1e-15)
    # Keep whichever point is more nearly stationary.
    psi = min((np.clip(candidate, floor, 1.0) for candidate in (search.x, polished.x)),
              key=lambda candidate: float(np.abs(gradient(candidate)).max()))
    if np.any(psi <= 10 * floor):
        raise ValueError("Heywood case: an item's uniqueness reached zero. "
                         "Use fewer factors or principal-axis extraction.")
    converged = float(np.abs(gradient(psi)).max()) < 1e-8
    return loadings_at(psi), int(search.nit), converged
```

```python
%%writefile redundant_variables.py
"""Which variables are exact linear combinations of the ones listed before them.

A variable that is a weighted sum of others adds no information and makes the
correlation matrix singular. No library names them, so it is written out: each
column, centred and scaled to unit length, is regressed on the columns kept so far,
and a residual sum of squares of 1e-10 or less marks it redundant. A constant column
is redundant too. The later member of a redundant pair is the one reported, so the
variable listed first survives.
"""

import numpy as np


def redundant_variables(x, names, tolerance=1e-10):
    centred = x - x.mean(axis=0)
    length = np.linalg.norm(centred, axis=0)
    length[length == 0] = 1.0
    centred = centred / length
    redundant, kept = [], []
    for j, name in enumerate(names):
        residual = centred[:, j]
        if kept:
            basis = centred[:, kept]
            residual = residual - basis @ np.linalg.lstsq(basis, residual, rcond=None)[0]
        if residual @ residual <= tolerance or x[:, j].max() == x[:, j].min():
            redundant.append(name)
        else:
            kept.append(j)
    return redundant
```

## Is the matrix worth factoring? KMO and Bartlett's test of sphericity

```python
# Bartlett's test: H0 is that the population correlation matrix is the identity, so a
# small P is the precondition for factoring. The statistic is
# -(n - 1 - (2p + 5)/6) ln|R| on p(p - 1)/2 degrees of freedom.
# KMO: squared correlations against squared partial correlations (read off R^-1),
# off the diagonal -- overall, and by variable (its column). Kaiser's (1974) words
# rate it. Both need R^-1 or ln|R|, which a collinear item left in takes away.
def kaiser(value):
    for bound, word in [(0.9, "Marvellous"), (0.8, "Meritorious"), (0.7, "Middling"),
                        (0.6, "Mediocre"), (0.5, "Miserable")]:
        if value >= bound:
            return word
    return "Unacceptable"


if any(name in variables for name in redundant):
    print("\nThe correlation matrix is singular: no KMO or Bartlett's test.")
else:
    chi_square_sphericity = -(n - 1 - (2 * p + 5) / 6) * np.linalg.slogdet(correlation)[1]
    df_sphericity = p * (p - 1) // 2
    sphericity = pd.DataFrame(
        [[chi_square_sphericity, df_sphericity, stats.chi2.sf(chi_square_sphericity, df_sphericity)]],
        columns=["chi-square", "df", "P"])
    inverse = np.linalg.inv(correlation)
    partial = -inverse / np.sqrt(np.outer(np.diag(inverse), np.diag(inverse)))
    off_diagonal = ~np.eye(p, dtype=bool)
    r2 = np.where(off_diagonal, correlation**2, 0.0).sum(axis=0)
    q2 = np.where(off_diagonal, partial**2, 0.0).sum(axis=0)
    kmo_overall = r2.sum() / (r2.sum() + q2.sum())
    kmo = pd.Series(r2 / (r2 + q2), index=variables)
    print("\nBartlett's test of sphericity")
    print(sphericity.to_string(index=False))
    print(f"\nKMO overall: {kmo_overall:.4f} ({kaiser(kmo_overall)})")
    print(pd.DataFrame({"KMO": kmo, "rating": kmo.map(kaiser)}).round(4))
```

## How many factors

```python
factors = 3
max_iterations = 50
```

## Starting communalities

```python
# The squared multiple correlation (SMC): the R-squared of regressing the item on all
# the others, 1 - 1 / (R inverse)_ii.
start = np.clip(1 - 1 / np.diag(np.linalg.pinv(correlation)), 0.0, 1 - 1e-8)
```

## Extracting the factors

```python
# Maximum likelihood: the loadings that make the observed correlations most likely if
# the items were multivariate normal. statsmodels' Factor(method="ml") fits the same
# model but stops about 1e-6 short of the optimum, so the fit is written out in
# factor_analysis_ml.py, by the profile likelihood over the uniquenesses (the variance
# the factors leave unexplained), starting from 1 - the starting communalities. A
# uniqueness that reaches zero is a Heywood case -- an item the factors would explain
# completely, which no real item is -- and the fit is refused, as StatNex refuses it.
loadings, iterations, converged = ml_loadings(correlation, factors, start, max_iterations)

# Factors in order of the variance they explain; a factor and its negative are the
# same factor, so each is flipped to make its loadings sum to a positive number.
loadings = loadings[:, np.argsort((loadings**2).sum(axis=0))[::-1]]
loadings = loadings * np.where(loadings.sum(axis=0) < 0, -1.0, 1.0)
communality = (loadings**2).sum(axis=1)
names = [f"Factor {j + 1}" for j in range(factors)]
settings = pd.Series({"factors": factors, "max iterations": max_iterations,
                      "iterations": iterations, "converged": converged})
print()
print(settings)
```

## Goodness-of-fit test

```python
# H0: this many factors reproduce the correlation matrix (up to sampling error), so a
# large P is the good outcome -- but it only fails to reject the model; it does not
# prove the number of factors right, and with a large n it rejects small misfits.
# Bartlett's multiplier n - 1 - (2p + 5)/6 - 2f/3 scales the discrepancy, as R's
# factanal() does; chi-square on ((p - f)^2 - p - f)/2 df.
# AIC = -2 log L + 2q, with -2 log L = (n - 1)(log|implied| + trace(implied^-1 R)) (the
# likelihood the fit maximises, up to a constant) and q the number of free parameters.
# When comparing numbers of factors on the same items, the smaller AIC is the better
# model.
implied = loadings @ loadings.T
np.fill_diagonal(implied, 1.0)
ratio = np.linalg.solve(implied, correlation)
discrepancy = np.trace(ratio) - np.linalg.slogdet(ratio)[1] - p
chi_square = (n - 1 - (2 * p + 5) / 6 - 2 * factors / 3) * discrepancy
df = ((p - factors) ** 2 - p - factors) // 2
parameters = p * factors + p - factors * (factors - 1) // 2
aic = (n - 1) * (np.linalg.slogdet(implied)[1] + np.trace(ratio)) + 2 * parameters
fit = pd.DataFrame([[aic, chi_square, df, stats.chi2.sf(chi_square, df)]],
                   columns=["AIC", "chi-square", "df", "P"])
print()
print(fit)
```

## Communalities

```python
communalities = pd.DataFrame({"start": start, "estimate": communality}, index=variables)
print()
print(communalities.round(4))
```

## Loadings before rotation

```python
print()
print(pd.DataFrame(loadings, index=variables, columns=names).round(4))
```

## Rotation

```python
# Rotation leaves the fit unchanged and only turns the axes so that each item loads
# on as few factors as possible. statsmodels' rotate_factors runs the gradient
# projection algorithm, here until the gradient norm is below 1e-10.
# Kaiser normalisation: each item's row is scaled to unit length first, so items with
# a high communality do not dominate, and scaled back afterwards.
weights = np.sqrt(communality)
weights[weights == 0] = 1.0
scaled = loadings / weights[:, None]
rotated = rotate_factors(scaled, "varimax", tol=1e-10, max_tries=2000)[0] * weights[:, None]
# Rotated factors are put in order and given a sign the same way as before.
rotated = rotated[:, np.argsort((rotated**2).sum(axis=0))[::-1]]
rotated = rotated * np.where(rotated.sum(axis=0) < 0, -1.0, 1.0)
final = rotated
print()
print(pd.DataFrame(rotated, index=variables, columns=names).round(4))
```

## Eigenvalues and variance explained

```python
# Initial: eigenvalues of R. Extracted: each factor's sum of squared loadings.
# Each is divided by the number of items for its share.
initial = np.sort(np.linalg.eigvalsh(correlation))[::-1]
extracted = (loadings**2).sum(axis=0)
eigen = np.full((p, 9), np.nan)
eigen[:, 0], eigen[:, 1], eigen[:, 2] = initial, initial / p, initial.cumsum() / p
eigen[:factors, 3], eigen[:factors, 4], eigen[:factors, 5] = extracted, extracted / p, extracted.cumsum() / p
# After rotation: the variance each factor accounts for, in the factors' order.
# The factors are uncorrelated, so it is each one's sum of squared loadings.
after = (final**2).sum(axis=0)
eigen[:factors, 6], eigen[:factors, 7], eigen[:factors, 8] = after, after / p, after.cumsum() / p
eigen = pd.DataFrame(eigen, index=range(1, p + 1),
                     columns=pd.MultiIndex.from_product([["initial", "extracted", "rotated"],
                                                         ["eigenvalue", "share", "cumulative"]]))
print()
print(eigen.round(4))
```

## What the result says

Factor analysis looks for a few hidden themes, called factors, behind many survey items. Here three factors were asked for, with varimax rotation, which turns the factors so that each item belongs mainly to one.

The data suit the method: the overall KMO measure, which is near 1 when the items overlap enough to share factors, is 0.82. The goodness-of-fit test gives a chi-square of 37.8 with P = 0.26. That P value is not small, so three factors are not shown to be too few. A loading is the correlation between an item and a factor. After rotation, items 05 to 08 load on factor 1 (0.59 to 0.86): service. Items 01 to 04 load on factor 2 (0.62 to 0.76 in size; item 04 is negative because it is worded the other way round): product. Items 09 to 12 load on factor 3 (0.59 to 0.69): the place.

So yes: the twelve items fall cleanly into the three intended themes. The survey cannot show which theme brings customers back.
