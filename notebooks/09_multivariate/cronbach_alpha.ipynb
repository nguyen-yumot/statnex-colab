## Helper: redundant_variables

A step that no library does, written out. In the code download it is the file `redundant_variables.py`, which the analysis imports; here the analysis uses it directly.

```python
"""Which variables are exact linear combinations of the ones listed before them.

A variable that is a weighted sum of others adds no information and makes the
correlation matrix singular. No library names them, so it is written out: each
column, centred and scaled to unit length, is regressed on the columns kept so far,
and a residual sum of squares of 1e-10 or less marks it redundant. A constant column
is redundant too. The later member of a redundant pair is the one reported, so the
variable listed first survives.
"""

import numpy as np


def redundant_variables(x, names, tolerance=1e-10):
    centred = x - x.mean(axis=0)
    length = np.linalg.norm(centred, axis=0)
    length[length == 0] = 1.0
    centred = centred / length
    redundant, kept = [], []
    for j, name in enumerate(names):
        residual = centred[:, j]
        if kept:
            basis = centred[:, kept]
            residual = residual - basis @ np.linalg.lstsq(basis, residual, rcond=None)[0]
        if residual @ residual <= tolerance or x[:, j].max() == x[:, j].min():
            redundant.append(name)
        else:
            kept.append(j)
    return redundant
```

## Cronbach's alpha

```python
# alpha = k / (k - 1) x (1 - sum of the item variances / variance of the total), for k
# items. It estimates the reliability of the total score: the share of its variance
# that is not measurement noise. It assumes every item measures the same one thing
# (to the same degree); it does not test that. It is high when the items rise and fall
# together, and it also rises with the number of items, so a long scale can score well
# with weak items. 0.7 is a common rule of thumb for a usable scale, not a law.
# Neither SciPy nor statsmodels has it, so the formula is written out; it is what R's
# psych::alpha and Stata's -alpha- report as the raw alpha.
def cronbach_alpha(x):
    k = x.shape[1]
    return k / (k - 1) * (1 - x.var(axis=0, ddof=1).sum() / x.sum(axis=1).var(ddof=1))


alpha = cronbach_alpha(x)
print(f"\nCronbach's alpha = {alpha:.4f}")
```

## If each item were deleted

```python
# For each item: the total of the remaining items (its mean and variance), the alpha
# without the item (with only one item left there is no alpha), the item's correlation
# with that remaining total (the "corrected" item-total correlation, so the item is not
# correlated with itself), and the R^2 of regressing the item on the other items (the
# squared multiple correlation, by statsmodels' OLS). An item whose removal raises the
# alpha, or whose corrected correlation is near zero or negative, does not measure what
# the others measure.
rows = []
for j, name in enumerate(items):
    rest = np.delete(x, j, axis=1)
    rest_total = rest.sum(axis=1)
    rows.append([
        rest_total.mean(),
        rest_total.var(ddof=1),
        cronbach_alpha(rest) if rest.shape[1] >= 2 else np.nan,
        np.corrcoef(x[:, j], rest_total)[0, 1],
        sm.OLS(x[:, j], sm.add_constant(rest)).fit().rsquared,
    ])
deleted = pd.DataFrame(rows, index=items, columns=["mean of total", "variance of total",
                                                   "alpha", "r with total", "SMC"])
print()
print(deleted)
```

## What the result says

Cronbach's alpha measures how well a set of survey items move together. It is close to 1 when the items measure one thing, and then they can be added into one score. The five-item service score reaches only 0.61, which is not reliable enough.

The table of the effect of deleting each variable shows why. Without item 09, about noise, alpha rises to 0.83. Deleting any one of the four staff items makes alpha fall instead, to between 0.40 and 0.48. Item 09 also has a negative correlation with the total of the other items, -0.24, so it points the opposite way.

So item 09 does not belong: it measures the place, not the staff. The four staff items on their own make a reliable service score.
