These cells save `collinear_columns.py`: a step that no library does, written out. The analysis imports it.

```python
%%writefile collinear_columns.py
"""Which explanatory variables are exact linear combinations of the ones before them.

Such a variable adds no information and makes the model impossible to estimate. Each
column is centred and scaled to length 1, then regressed on the columns already kept;
if nothing is left over (squared residual length at most 1e-10) the column is
redundant. A constant column is redundant too. The *later* member of a redundant pair
is the one reported, so a variable listed first survives.
"""

import numpy as np


def collinear_columns(values, names, tolerance=1e-10):
    centred = values - values.mean(axis=0)
    scale = np.linalg.norm(centred, axis=0)
    scale[scale == 0] = 1.0
    centred = centred / scale
    dropped, kept = [], []
    for j, name in enumerate(names):
        column = centred[:, j]
        if kept:
            basis = centred[:, kept]
            column = column - basis @ np.linalg.lstsq(basis, column, rcond=None)[0]
        if column @ column <= tolerance or values[:, j].max() == values[:, j].min():
            dropped.append(name)
        else:
            kept.append(j)
    return dropped
```

## Libraries

```python
import re
import warnings

import numpy as np
import pandas as pd
from scipy import stats
from statsmodels.miscmodels.ordinal_model import OrderedModel
from statsmodels.tools.sm_exceptions import ConvergenceWarning
from collinear_columns import collinear_columns

pd.set_option("display.width", 160)
pd.set_option("display.max_columns", 20)
```

## The data

```python
# Text in an explanatory column counts as missing.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
response = "Likely to recommend"
predictors = ["Loyalty member", "Uses app", "Satisfaction score"]
frame = data[predictors].apply(pd.to_numeric, errors="coerce")
max_iter = 50
confidence = 95 / 100
```

## Cases

```python
y_ok = data[response].notna()
x_ok = frame.notna().all(axis=1)
summary = pd.DataFrame(
    {"n": [(y_ok & x_ok).sum(), (~y_ok & x_ok).sum(), (y_ok & ~x_ok).sum(),
           (~y_ok & ~x_ok).sum(), len(frame)]},
    index=["valid", "response missing", "explanatory missing", "both missing", "total"],
)
summary["%"] = summary["n"] / len(frame)
print(summary)
valid = frame[y_ok & x_ok]
outcome = data.loc[y_ok & x_ok, response].to_numpy()
n = len(outcome)

# The categories in numeric order, the smallest first. Text has no order of its own:
# sorted as text, Low/Medium/High would become High < Low < Medium, so for text the app
# asks for the order in "Order of the categories".
levels = sorted(set(outcome), key=float)
codes = np.array([levels.index(v) for v in outcome])
counts = np.bincount(codes, minlength=len(levels))
response_summary = pd.DataFrame({"n": [*counts, n]}, index=[*map(str, levels), "total"])
response_summary["%"] = response_summary["n"] / n
print()
print(response_summary)
```

## Descriptive statistics and correlations, overall and by category

```python
groups = {"all": valid, **{str(level): valid[codes == k] for k, level in enumerate(levels)}}
descriptive = pd.concat({
    name: pd.DataFrame({"n": len(part), "mean": part.mean(), "variance": part.var(ddof=1),
                        "SD": part.std(ddof=1), "min": part.min(), "max": part.max()})
    for name, part in groups.items()
})
correlation = pd.concat({name: part.corr() for name, part in groups.items()})
print()
print(descriptive)
print()
print(correlation)
```

## Variables that are linear combinations of others

```python
values = valid.to_numpy(float)
dropped = collinear_columns(values, predictors)
print("\nLinearly dependent:", dropped or "none")
# They are left out of the model.
predictors = [name for name in predictors if name not in dropped]
values = valid[predictors].to_numpy(float)
```

## The proportional-odds model

```python
# logit P(Y <= k) = threshold_k - x'b: one set of slopes shared by every threshold.
# BFGS run to a gradient of 1e-12, as in the app (statsmodels' default stops ~1e-5 short).
# A gradient of 1e-12 is more than double precision can always reach, so BFGS may stop a
# hair short with "precision loss" and statsmodels would warn that it failed. Judge the
# stop instead by the gradient it reached, at statsmodels' own tolerance (1e-5), as the
# app does: only a real failure is reported.
with warnings.catch_warnings():
    warnings.simplefilter("ignore", ConvergenceWarning)
    model = OrderedModel(codes, values, distr="logit").fit(
        method="bfgs", disp=0, maxiter=max(max_iter, 500), gtol=1e-12)
if np.abs(model.model.score(model.params)).max() > 1e-5:
    print("Warning: the fit has not converged; read the results as provisional.")
deviance = -2 * model.llf
# With no variable, each category's probability is its share of the cases.
null_deviance = -2 * (counts * np.log(counts / n)).sum()
slopes, thresholds = len(predictors), len(levels) - 1

# Pseudo R²: Cox-Snell, Nagelkerke and McFadden.
cox_snell = 1 - np.exp(-(null_deviance - deviance) / n)
accuracy = pd.DataFrame([[cox_snell, cox_snell / (1 - np.exp(-null_deviance / n)),
                          1 - deviance / null_deviance]],
                        columns=["Cox-Snell", "Nagelkerke", "McFadden"])
chi2 = null_deviance - deviance
model_tests = pd.DataFrame(
    [[null_deviance, null_deviance + 2 * thresholds, np.nan, np.nan, np.nan],
     [deviance, deviance + 2 * (slopes + thresholds), chi2, slopes, stats.chi2.sf(chi2, slopes)]],
    index=["without the variables", "with the variables"],
    columns=["-2LL", "AIC", "chi-square", "df", "P"])
print()
print(accuracy.to_string(index=False))
print()
print(model_tests)
```

## Coefficients

```python
# statsmodels stores the thresholds as the first one plus log increments; the delta method
# turns that covariance into standard errors of the thresholds themselves. The app writes
# the model as logit P(Y <= k) = b0k + ..., so its constant is minus the threshold.
params, covariance = np.asarray(model.params), np.asarray(model.cov_params())
raw = params[slopes:]
cut_points = np.concatenate([[raw[0]], raw[0] + np.cumsum(np.exp(raw[1:]))])
jacobian = np.zeros((thresholds, thresholds))
for k in range(thresholds):
    jacobian[k, 0] = 1.0
    jacobian[k, 1:k + 1] = np.exp(raw[1:k + 1])
cut_se = np.sqrt(np.diag(jacobian @ covariance[slopes:, slopes:] @ jacobian.T))
b = np.concatenate([params[:slopes], -cut_points])
se = np.concatenate([np.asarray(model.bse)[:slopes], cut_se])
z = stats.norm.ppf(0.5 + confidence / 2)
coefficients = pd.DataFrame({
    "b": b, "SE": se, "lower": b - z * se, "upper": b + z * se,
    "odds ratio": np.exp(b), "OR lower": np.exp(b - z * se), "OR upper": np.exp(b + z * se),
    "Wald": (b / se) ** 2, "df": 1, "P": stats.chi2.sf((b / se) ** 2, 1),
}, index=[*predictors, *[f"constant {response}={level}" for level in levels[:-1]]])
print()
print(coefficients)
```

## Classification table

```python
# Each case is assigned to its most probable category.
probabilities = model.predict()
predicted = probabilities.argmax(axis=1)
classification = pd.crosstab(codes, predicted).reindex(
    index=range(len(levels)), columns=range(len(levels)), fill_value=0)
classification.index = classification.columns = list(map(str, levels))
classification["hit rate"] = np.diag(classification) / classification.sum(axis=1)
hit_rate = (predicted == codes).mean()
print()
print(classification)
print(f"overall hit rate: {hit_rate:.4f}")
```

## What the result says

Ordinal logistic regression fits an outcome whose answers have an order, here unlikely, maybe and very likely to recommend. One odds ratio for each predictor says how the odds of a higher answer multiply when that predictor rises by one.

All three predictors are significant: their P values are almost zero, so chance alone would rarely give such links. Satisfaction has the largest odds ratio: 4.64 for each extra point, with a confidence interval (the range of values the data fit) from 3.22 to 6.69. Loyalty members have 2.75 times the odds of non-members, and app users 2.37 times. The classification table shows the limits: the model puts 56% of the 400 customers on the right step.

So all three push customers up the ladder, and satisfaction pushes hardest. But this does not show that joining or installing the app makes anyone recommend the shop.
