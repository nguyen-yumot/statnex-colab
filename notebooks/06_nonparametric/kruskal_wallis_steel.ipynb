# Listing photos and views: Steel's test against 1 photo

**Kruskal-Wallis Test with Multiple Comparisons** · a worked example from the StatNex teaching guide

**The story.** A resale app looked at 21 listings for the same kind of jacket, posted with 1, 3, 5 or 10 photos, and counted how many people viewed each listing in its first week.

**The question.** Do listings with more photos get more views, and how many photos are enough?

## Libraries

```python
import numpy as np
import pandas as pd
from scipy import stats
from scipy import integrate

pd.set_option("display.width", 120)
```

## The data

```python
# One column per level. Text and blank cells are skipped, so levels may differ in size.
data = pd.read_csv("data.csv", keep_default_na=False, na_values=[""])
levels = ["1 photo", "3 photos", "5 photos", "10 photos"]
groups = [pd.to_numeric(data[name], errors="coerce").dropna().to_numpy(float) for name in levels]
sizes = np.array([len(g) for g in groups])
```

## Summary by level

```python
# All values are ranked together (ties share the average rank); each level's mean rank
# is what the test compares.
ranks = stats.rankdata(np.concatenate(groups))
mean_ranks = [part.mean() for part in np.split(ranks, np.cumsum(sizes)[:-1])]
summary = pd.DataFrame({"n": sizes, "mean rank": mean_ranks}, index=levels).T
print(summary)
```

## Kruskal-Wallis test

```python
# SciPy's H carries the tie correction; it is compared with chi-square on k - 1 df.
h, p = stats.kruskal(*groups)
df = len(groups) - 1
print(f"\nchi-square = {h:.6f}, df = {df}, P = {p:.6g}")


# Wilcoxon's rank-sum statistic for one pair, ranked on its own (not with the other
# levels), standardised with the tie-corrected variance. Positive when the second
# group sits higher.
def rank_sum_z(a, b):
    combined = np.concatenate([a, b])
    total = len(combined)
    r = stats.rankdata(combined)
    expected = len(b) * (total + 1) / 2
    variance = len(a) * len(b) / (total * (total - 1)) * ((r**2).sum() - total * (total + 1) ** 2 / 4)
    return (r[len(a):].sum() - expected) / np.sqrt(variance)
```

## Multiple comparison: Steel

```python
# Each level against the control -- the first level here -- each pair ranked on its own.
control = 0
# The statistics share the control, so they are correlated: corr(z_i, z_j) =
# sqrt(l_i l_j) with l_i = n_i / (n0 + n_i + 1), Steel's finite-sample value (Dunnett's
# n_i / (n0 + n_i) is its large-sample limit). P is the chance that the largest |z|
# is at least this one -- Dunnett's distribution at infinite df -- which that
# one-factor structure turns into a single integral over the control's share.
others = [i for i in range(len(groups)) if i != control]
share = np.array([sizes[i] / (sizes[control] + sizes[i] + 1) for i in others])


def steel_p(z):
    def outside(u):
        low = (-abs(z) - np.sqrt(share) * u) / np.sqrt(1 - share)
        high = (abs(z) - np.sqrt(share) * u) / np.sqrt(1 - share)
        return stats.norm.pdf(u) * (1 - np.prod(stats.norm.cdf(high) - stats.norm.cdf(low)))
    return integrate.quad(outside, -np.inf, np.inf, epsabs=1e-14, epsrel=1e-12)[0]


rows = []
for i in others:
    z = rank_sum_z(groups[control], groups[i])
    rows.append([levels[control], levels[i], z, steel_p(z)])
pairs = pd.DataFrame(rows, columns=["level 1", "level 2", "z", "P"])
print()
print(pairs.to_string(index=False))
```

## What the result says

The Kruskal-Wallis test compares groups by ranks, the positions of the listings when sorted by views. Its chi-square statistic is 14.89 and the P value is 0.002. The P value is the chance of gaps this large if photos made no difference, so the photo counts differ somewhere.

Steel's test compares each photo count only with the control group, here 1 photo. That makes three comparisons, not all six pairs, so the test gives up less to guard against false alarms. Five photos beat one (P = 0.011) and ten photos beat one (P = 0.018). Three photos against one give P = 0.143, which is not significant.

Five or ten photos clearly get more views than one photo. A gain from three photos is not shown, which is not proof of no gain. This test cannot say whether ten photos beat five.
